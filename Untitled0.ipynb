# Task
Develop an offline AI-powered Flutter application for Android that includes a chatbot feature, text-to-image generation, and image modification (uploaded image + prompt). This involves researching and integrating suitable offline AI models, setting up a Flutter project, implementing a clean UI/UX, and optimizing for performance on Android devices.

## Research and AI Model Selection

### Subtask:
Research and select suitable AI models for offline chatbot and image generation (text-to-image and image-to-image/modification). This includes considering models compatible with on-device deployment (e.g., TensorFlow Lite, PyTorch Mobile) for offline functionality, and evaluating their performance, size, and capabilities.


### Research on On-Device AI Deployment Frameworks: TensorFlow Lite and PyTorch Mobile

To enable offline functionality for the AI chatbot and image generation features on Android devices, it's crucial to select appropriate on-device AI deployment frameworks. The primary candidates are TensorFlow Lite and PyTorch Mobile, both designed to optimize machine learning models for mobile and edge devices.

#### TensorFlow Lite (TFLite)

TensorFlow Lite is Google's lightweight library for deploying TensorFlow models on mobile, embedded, and IoT devices. Its key characteristics include:

*   **Optimized Binary Size**: TFLite models (`.tflite` format) are highly optimized for size, making them suitable for devices with limited storage.
*   **Low Latency**: It employs various optimizations, including quantizing models (reducing precision of weights and activations) and pre-fused operations, to achieve fast inference speeds.
*   **Cross-platform**: Supports Android, iOS, Linux, and microcontrollers.
*   **Interpreter**: Features a compact interpreter for executing TFLite models efficiently.
*   **Model Conversion**: Offers a converter tool to transform existing TensorFlow models into the TFLite format. This includes support for many common layers and operations.
*   **Hardware Acceleration**: Can leverage hardware accelerators (e.g., Android Neural Networks API, GPU delegate) for further performance improvements.
*   **Strengths**: Mature ecosystem, extensive documentation, good support for quantization, strong community backing, and integration with Android Studio.
*   **Limitations**: Can be restrictive in terms of custom operations; sometimes requires careful model design to ensure full compatibility and optimal performance without custom ops.

#### PyTorch Mobile

PyTorch Mobile is the mobile solution from Facebook's PyTorch framework, designed to bring PyTorch models to iOS and Android devices. Key features include:

*   **End-to-end Workflow**: Provides a streamlined workflow from model training in PyTorch to deployment on mobile.
*   **TorchScript**: Uses TorchScript, a subset of Python, to serialize models. This allows models to be run in a non-Python environment and provides optimizations for mobile deployment.
*   **Optimized Runtime**: Offers a compact runtime that supports a wide range of PyTorch operators.
*   **Quantization Support**: Includes post-training quantization and quantization-aware training to reduce model size and improve inference speed.
*   **Custom Operators**: Good support for custom C++ operators, offering flexibility for complex models or unique requirements.
*   **Hardware Acceleration**: Integrates with mobile device's native capabilities for accelerated inference.
*   **Strengths**: High flexibility for custom models and operations, strong integration with the PyTorch ecosystem (which is popular for research), and good support for dynamic graph execution if needed.
*   **Limitations**: Can sometimes result in larger binary sizes compared to TFLite for similar models, and its ecosystem for mobile deployment is somewhat less mature than TFLite's.

#### Initial Conclusion

Both TensorFlow Lite and PyTorch Mobile are strong contenders for on-device AI deployment. For this project, given the emphasis on a 'lite end app' and potentially limited device resources, **TensorFlow Lite appears to be a slightly more advantageous starting point, especially for models that can benefit significantly from aggressive quantization and whose operations are well-covered by TFLite's optimized set.** Its deeper integration and established presence within the Android ecosystem also provide a robust foundation. However, PyTorch Mobile offers compelling flexibility, particularly if custom operations or a more dynamic model structure become necessary.

We will proceed with investigating model compatibility primarily focusing on TensorFlow Lite, while keeping PyTorch Mobile as a viable alternative, especially if TFLite poses significant limitations for specific model architectures or functionalities.

1.  **Model Size**: Critical for on-device deployment. Smaller models mean faster download times, less storage space, and quicker loading into memory.

3.  **Accuracy/Performance**: The model must provide meaningful and relevant responses, even if it means a slightly simpler dialogue system compared to cloud-based solutions.
4.  **Offline Capability**: Must function entirely without an internet connection.
5.  **Conversion/Compatibility**: Ease of conversion to TFLite or PyTorch Mobile format, and compatibility with their respective runtimes.
6.  **Licensing**: Open-source licenses compatible with commercial use are preferred.

    *   **Description**: Recurrent Neural Networks (RNNs) like LSTMs or GRUs, often combined with word embeddings (e.g., Word2Vec, GloVe), can be very effective for intent classification.

### Research on AI Models for Offline Text-to-Image Generation

For offline text-to-image generation, the challenge lies in deploying complex generative models on resource-constrained mobile devices. The goal is to identify models that can produce reasonable quality images from text prompts while adhering to strict limitations on model size and inference speed. We will primarily focus on models that can be converted to TensorFlow Lite (TFLite) or PyTorch Mobile.

#### Key Considerations for Text-to-Image Models:

1.  **Model Size**: Generative models, especially diffusion models, can be very large. Significant distillation, pruning, or quantization is often required to make them feasible for on-device deployment.
2.  **Inference Speed**: Image generation can be computationally intensive. Real-time or near real-time generation is desirable, but a few seconds might be acceptable depending on the complexity of the output.
3.  **Image Quality**: While not expecting state-of-the-art cloud-level quality, the generated images should be visually coherent and relevant to the text prompt.
4.  **Offline Capability**: Must function entirely without an internet connection.
5.  **Conversion/Compatibility**: Ease of conversion to TFLite or PyTorch Mobile format. This often involves ensuring that all model operations are supported by the respective mobile runtime.
6.  **Licensing**: Open-source licenses compatible with commercial use are preferred.

#### Candidate Model Categories and Examples:

##### 1. Distilled/Quantized Diffusion Models

Diffusion models like Stable Diffusion are powerful but very large. Distillation techniques can create smaller, faster versions.

*   **Mini/Tiny versions of Stable Diffusion (or similar diffusion models)**:
    *   **Description**: Researchers are actively working on creating smaller, more efficient versions of popular diffusion models. These often involve reducing the number of layers, using smaller latent spaces, or employing knowledge distillation.
    *   **TFLite/PyTorch Mobile Compatibility**: Challenging due to the architecture complexity and the custom operations often involved. Requires significant effort in model optimization (e.g., pruning, quantization, custom operator implementation).
    *   **Pros**: Potential for high-quality image generation, even in distilled forms, as diffusion models are highly effective.
    *   **Cons**: Still likely to be large and slow compared to other approaches. Conversion to mobile formats can be highly complex and may result in noticeable quality degradation.

##### 2. Generative Adversarial Networks (GANs)

GANs have been used for image generation for a longer time and some architectures are more amenable to smaller sizes.

*   **"Tiny" GANs (e.g., GANs specifically designed for edge devices, or distilled versions)**:
    *   **Description**: Architectures like BigGAN or StyleGAN are too large, but simpler GAN variants or heavily compressed versions (e.g., MobileGAN) can generate images. Often focus on specific domains (e.g., faces, simple objects).
    *   **TFLite/PyTorch Mobile Compatibility**: More feasible than large diffusion models for conversion, especially if they use standard convolutional layers. Quantization can further reduce size.
    *   **Pros**: Can be smaller and faster than diffusion models. Some architectures are well-understood for mobile deployment.
    *   **Cons**: Image quality typically lags behind state-of-the-art diffusion models. Can be prone to mode collapse or generating less diverse outputs.

*   **DALL-E mini (Craiyon) inspired models**:
    *   **Description**: While DALL-E mini itself might be too large, its architecture, which often involves a VQ-VAE for image generation combined with a Transformer for text-to-latent mapping, could be simplified or distilled.
    *   **TFLite/PyTorch Mobile Compatibility**: The VQ-VAE part might be convertible, but the large Transformer component for text encoding and latent generation poses significant challenges for mobile deployment due to its size and computational demands.
    *   **Pros**: Can generate diverse images from text.
    *   **Cons**: Likely too large and slow for practical on-device use, even after significant optimization.

#### Initial Recommendation for Text-to-Image Generation

Given the extreme constraints of offline mobile deployment for a 'lite end app', **on-device text-to-image generation is currently the most challenging feature to implement with satisfactory performance and quality.**

1.  **Prioritize Small Diffusion Models**: The most promising path forward is to investigate the latest advancements in **highly distilled or compact versions of diffusion models (e.g., Tiny Diffusion Models, Mobile Diffusion models)**. While these are still emerging, they offer the best balance of potential quality and the architectural foundation for size reduction.
2.  **Explore Optimized GANs for Specific Domains**: If a very specific style or domain of image generation is acceptable, **highly optimized and compact GAN architectures** could be a viable alternative due to their potentially smaller size and faster inference.
3.  **Acknowledge Limitations**: It's crucial to set realistic expectations for the quality and complexity of images that can be generated entirely offline on a mobile device. Significant compromises on output quality or generation speed may be necessary.

We will actively search for pre-trained models or existing research focused on extremely lightweight diffusion models or GANs that have demonstrated successful deployment on mobile platforms, with a strong preference for TFLite compatible solutions.

### Research on AI Models for Offline Image-to-Image/Modification

For offline image modification, the goal is to find models that can take an input image and a text prompt, then modify the image according to the prompt, all while running efficiently on mobile devices. This involves aspects of image-to-image translation, conditional image generation, or style transfer. As with other features, model size, inference speed, and compatibility with TensorFlow Lite (TFLite) or PyTorch Mobile are paramount.

#### Key Considerations for Image Modification Models:

1.  **Model Size**: These models can be complex, requiring significant optimization (distillation, pruning, quantization) to fit on mobile devices.
2.  **Inference Speed**: Modifications should ideally be near real-time, but a few seconds might be acceptable for more complex transformations.
3.  **Quality of Modification**: The output image should clearly reflect the requested modification while maintaining visual coherence and the integrity of the original image's relevant features.
4.  **Offline Capability**: Must operate fully without internet access.
5.  **Conversion/Compatibility**: Ease of conversion to TFLite or PyTorch Mobile, ensuring all necessary operations are supported by the mobile runtime.
6.  **Licensing**: Open-source licenses compatible with commercial use are preferred.
7.  **Input Modalities**: The model must effectively handle both image and text inputs to guide the modification.

#### Candidate Model Categories and Examples:

##### 1. Conditional Diffusion Models for Image-to-Image

Advanced diffusion models can perform image editing conditioned on text and an input image.

*   **Img2Img Diffusion Models (e.g., controlnet-like, pix2pix-zero variants)**:
    *   **Description**: These models use an input image along with a text prompt to guide the diffusion process for editing. They are highly effective for tasks like changing object styles, adding/removing elements, or transforming images based on descriptions.
    *   **TFLite/PyTorch Mobile Compatibility**: Extremely challenging. Even distilled versions of text-to-image diffusion models are difficult to deploy. Adding an image input and the complexity of preserving content while modifying it makes these models very large and computationally intensive. Requires extensive research into highly specialized mobile-optimized architectures or significant quantization/pruning.
    *   **Pros**: High fidelity and creative control over modifications.
    *   **Cons**: Currently, very difficult to run efficiently and effectively on-device due to their size and computational demands. Likely too slow and large for a 'lite end app' in a truly offline scenario.

##### 2. Style Transfer Networks

For modifying the artistic style of an image based on a prompt or a reference style.

*   **Arbitrary Style Transfer Networks (e.g., Mobile-optimized CNNs or GANs for style transfer)**:
    *   **Description**: These models can apply a new style (often learned from a large dataset or a reference image) to a content image. Text prompts could be used to select a style from a pre-defined set or to guide a simpler, more abstract style change.
    *   **TFLite/PyTorch Mobile Compatibility**: More feasible than full diffusion models. Some CNN-based style transfer models have been successfully optimized for mobile (e.g., using quantization or smaller architectures).
    *   **Pros**: Can achieve good aesthetic results for style modification. Relatively faster inference compared to generative models.
    *   **Cons**: Limited to style changes; less flexible for arbitrary object modifications or content-aware editing based on text prompts. The text-to-style mapping might be simplistic or require pre-defined styles.

##### 3. Pix2Pix or CycleGAN Variants

These are image-to-image translation models, typically for specific domain translations (e.g., day to night, sketches to photos). A text prompt could select the target domain or specific attribute.

*   **Lightweight Pix2Pix/CycleGAN models**:
    *   **Description**: These GAN-based models learn a mapping from an input image to an output image in a target domain. To incorporate text, the text prompt would need to condition the generation process or select between multiple pre-trained domain translation models.
    *   **TFLite/PyTorch Mobile Compatibility**: Possible for specific, simple transformations after significant architectural simplification, pruning, and quantization. The models are usually smaller than diffusion models.
    *   **Pros**: Effective for well-defined, learned image-to-image transformations.
    *   **Cons**: Training for each specific modification task can be extensive. Integrating text conditioning effectively while keeping the model lightweight is a significant challenge. Generalization to novel modification requests is limited.

#### Initial Recommendation for Image Modification

Similar to text-to-image generation, **on-device image modification guided by text prompts presents significant challenges for a 'lite end app' due to the computational intensity and size of state-of-the-art models.**

1.  **Prioritize Simplified Style Transfer or Domain Translation**: The most pragmatic approach would be to focus on **highly optimized and simplified style transfer networks** or **lightweight GAN-based image-to-image translation models** for very specific, pre-defined modification tasks. The text prompt would likely act as a selector for these pre-defined operations or styles, rather than driving highly nuanced, arbitrary content modifications.
2.  **Explore Hybrid Approaches with Server-Side Fallback**: If true arbitrary text-guided image modification is a critical feature and on-device performance proves insufficient, a hybrid approach might be necessary where simpler modifications are done offline, and more complex ones are offloaded to a server (if an internet connection is available). This, however, goes against the strict 'offline functionality' requirement for this subtask.
3.  **Set Realistic Expectations**: It's crucial to manage expectations for the complexity and quality of image modifications achievable entirely offline on a mobile device. The focus should be on robust, performant basic modifications rather than highly generative or content-aware editing.

We will actively investigate research into extremely lightweight image-to-image models that incorporate textual conditioning, with a preference for those demonstrating TFLite compatibility and practical on-device performance.

### Consolidated Findings and Model Selection

Based on the detailed research into on-device AI deployment frameworks and candidate models for each feature, we can now consolidate our findings and make initial selections, keeping in mind the 'lite end app' requirement for offline functionality on Android devices.

#### 1. On-Device AI Deployment Framework:

*   **Selected Framework**: **TensorFlow Lite (TFLite)**
*   **Rationale**: TFLite offers a more mature ecosystem, stronger integration with Android development, robust support for aggressive quantization, and generally results in smaller model footprints and faster inference speeds compared to PyTorch Mobile for models that fit within its optimized operation set. This aligns well with the 'lite end app' and limited device resources. PyTorch Mobile remains a viable alternative if specific model architectures or custom operations prove challenging with TFLite.

#### 2. AI Model for Offline Chatbot:

*   **Selected Approach**: **Hybrid Model (Lightweight NLU + Retrieval-Based Response Generation)**
*   **Rationale**:
    *   **NLU (Intent Recognition & Entity Extraction)**: A **lightweight BERT-variant (e.g., MobileBERT or TinyBERT) or a highly optimized LSTM/GRU-based model** will be used for core Natural Language Understanding. These models offer good accuracy for NLU tasks while being more amenable to TFLite conversion and on-device execution. They are crucial for understanding user intent and extracting key information efficiently.
    *   **Response Generation**: A **retrieval-based system** is recommended for generating responses. This allows for controlled, consistent, and pre-defined answers, significantly reducing the model size and complexity compared to generative language models. The NLU component will handle the matching of user input to the most appropriate pre-written response. This approach prioritizes reliability, efficiency, and manageability of model size for offline mobile deployment.
*   **Evaluation against criteria**:
    *   **Offline Compatibility**: Excellent, as NLU models convert well to TFLite and retrieval is logic-based.

### Establish Clean Architecture: Initial Folder Structure

To ensure maintainability, scalability, and efficient UI development, we will establish a clean architecture for the Flutter application. This architecture separates the application into distinct layers, promoting separation of concerns.

We will create the following core directories inside `ai_chatbot_app/lib`:

*   **`data`**: This layer is responsible for data retrieval and manipulation. It defines and implements repositories, data sources (e.g., local storage, remote APIs, local AI models), and data transfer objects (DTOs).
    *   `data/datasources`: For defining abstract and concrete data sources (e.g., `ai_remote_datasource.dart`, `local_storage_datasource.dart`).
    *   `data/models`: For data models used internally by the data layer (often converted from DTOs or used for local storage).
    *   `data/repositories`: For concrete implementations of domain-layer repository interfaces.

*   **`domain`**: This is the core business logic layer. It is independent of any frameworks or external libraries. It defines entities, use cases (interactors), and repository interfaces.
    *   `domain/entities`: For core business objects.
    *   `domain/repositories`: For abstract repository interfaces that define how data is accessed.
    *   `domain/usecases`: For specific business operations or workflows.

*   **`presentation`**: This layer is responsible for the user interface and handling user interactions. It includes widgets, views, and state management logic (e.g., using `provider` as selected).
    *   `presentation/pages`: For top-level screens or views of the application.
    *   `presentation/widgets`: For reusable UI components.
    *   `presentation/providers`: For state management logic using the `provider` package (or view models/controllers).

*   **`core`**: This directory will contain common functionalities, utilities, constants, and extensions that are used across multiple layers.
    *   `core/utils`: For helper functions, formatters, etc.
    *   `core/constants`: For application-wide constants.

This structure provides a clear separation of responsibilities, making the application easier to understand, test, and modify.

**Reasoning**:
Based on the previously defined clean architecture, I will now create the necessary directories and subdirectories within the Flutter project's `lib` folder.

````python
import os

pubspec_path = os.path.join('ai_chatbot_app', 'pubspec.yaml')

# Read the existing pubspec.yaml file
with open(pubspec_path, 'r') as f:
    pubspec_content = f.readlines()

# Find the 'flutter:' section and then the 'assets:' subsection
flutter_assets_start_index = -1
for i, line in enumerate(pubspec_content):
    if line.strip() == 'flutter:':
        # Look for 'assets:' after 'flutter:'
        for j in range(i + 1, len(pubspec_content)):
            if pubspec_content[j].strip().startswith('assets:'):
                flutter_assets_start_index = j
                break
        if flutter_assets_start_index != -1:
            break

if flutter_assets_start_index == -1:
    print("Error: 'flutter: assets:' section not found in pubspec.yaml. Manually update or verify structure.")
else:
    # Check if assets/models/ is already declared
    asset_already_declared = False
    for i in range(flutter_assets_start_index + 1, len(pubspec_content)):
        # Stop if we hit another top-level section (e.g., 'environment:', 'dependencies:')
        if not pubspec_content[i].startswith('  -') and not pubspec_content[i].strip() == '':
            break
        if 'assets/models/' in pubspec_content[i]:
            asset_already_declared = True
            break

    if not asset_already_declared:
        # Insert '  - assets/models/' after the 'assets:' line
        pubspec_content.insert(flutter_assets_start_index + 1, '    - assets/models/\n')
        # Write the modified content back to pubspec.yaml
        with open(pubspec_path, 'w') as f:
            f.writelines(pubspec_content)
        print("Added 'assets/models/' to pubspec.yaml. Running flutter pub get...")
        os.chdir('ai_chatbot_app')
        !flutter pub get
        os.chdir('..')
        print("flutter pub get completed.")
    else:
        print("'assets/models/' already declared in pubspec.yaml. No changes made.")
````

**Reasoning**:
The previous code failed because it couldn't find the 'assets:' section in 'pubspec.yaml', likely due to it being commented out or missing. The updated code provides a more robust way to find the 'flutter:' section, then locate or create an 'assets:' subsection (uncommenting it if necessary) and then insert the 'assets/models/' path, ensuring proper formatting.



````python
import os

pubspec_path = os.path.join('ai_chatbot_app', 'pubspec.yaml')

# Read the existing pubspec.yaml file
with open(pubspec_path, 'r') as f:
    pubspec_content = f.readlines()

flutter_section_start_index = -1
assets_section_index = -1

# Find the 'flutter:' section
for i, line in enumerate(pubspec_content):
    if line.strip() == 'flutter:':
        flutter_section_start_index = i
        break

if flutter_section_start_index == -1:
    print("Error: 'flutter:' section not found in pubspec.yaml. Manually update or verify structure.")
else:
    # Search for 'assets:' within the flutter section
    for i in range(flutter_section_start_index + 1, len(pubspec_content)):
        line = pubspec_content[i]
        stripped_line = line.strip()

        # Stop if we hit another top-level key or a key with the same indentation as 'flutter:'
        if not line.startswith(' ') and stripped_line != '' and line.find(':') > -1:
            break
        # If we encounter a line with less indentation than '  assets:', it's another main section
        if line.startswith(' ') and line.find(':') > -1 and len(line) - len(line.lstrip()) < 2:
            break

        if stripped_line == 'assets:':
            assets_section_index = i
            break
        elif stripped_line == '# assets:': # Handle commented assets
            pubspec_content[i] = line.replace('# assets:', '  assets:') # Uncomment and ensure indentation
            assets_section_index = i
            break

    # If 'assets:' section was not found, insert it
    if assets_section_index == -1:
        insert_point = flutter_section_start_index + 1 # Default insert point
        # Find a suitable place to insert '  assets:' (e.g., after 'uses-material-design: true')
        for i in range(flutter_section_start_index + 1, len(pubspec_content)):
            if pubspec_content[i].strip() == 'uses-material-design: true':
                insert_point = i + 1
                break
            # If we find another non-empty line at a higher level of indentation, insert before it.
            if pubspec_content[i].strip() != '' and not pubspec_content[i].startswith('  '):
                insert_point = i
                break

        # Ensure there's a newline before inserting if the line above is not empty and doesn't end with newline
        if insert_point > 0 and not pubspec_content[insert_point-1].endswith('\n'):
             pubspec_content.insert(insert_point, '\n')
             insert_point += 1

        pubspec_content.insert(insert_point, '  assets:\n')
        assets_section_index = insert_point # Update index for subsequent insertion

    # Now that assets_section_index is identified/created, add 'assets/models/'
    asset_to_add = '    - assets/models/\n'
    asset_already_declared = False

    # Check if 'assets/models/' is already declared under the assets section
    # Iterate from after assets_section_index until another major section or less indented line
    for i in range(assets_section_index + 1, len(pubspec_content)):
        line = pubspec_content[i]
        # Stop if we hit a line with less indentation than '    - '
        if line.strip() != '' and not line.startswith('    '):
            break
        if asset_to_add.strip() == line.strip():
            asset_already_declared = True
            break

    if not asset_already_declared:
        # Find the correct insertion point immediately after the last existing asset or the 'assets:' line itself
        insert_asset_line_index = assets_section_index + 1
        while insert_asset_line_index < len(pubspec_content) and \
              (pubspec_content[insert_asset_line_index].strip() == '' or \
               pubspec_content[insert_asset_line_index].startswith('    -')):
            insert_asset_line_index += 1

        pubspec_content.insert(insert_asset_line_index, asset_to_add)

        # Write the modified content back to pubspec.yaml
        with open(pubspec_path, 'w') as f:
            f.writelines(pubspec_content)
        print("Added 'assets/models/' to pubspec.yaml. Running flutter pub get...")
        os.chdir('ai_chatbot_app')
        !flutter pub get
        os.chdir('..')
        print("flutter pub get completed.")
    else:
        print("'assets/models/' already declared in pubspec.yaml. No changes made.")
````

**Reasoning**:
The previous code failed because it was modifying the `flutter:` section under `dependencies:` instead of the top-level `flutter:` section for assets. I will fix this by ensuring the code specifically targets the top-level `flutter:` entry and correctly inserts or uncomments the `assets:` section with proper indentation, before adding the `assets/models/` path.



````python
import os

pubspec_path = os.path.join('ai_chatbot_app', 'pubspec.yaml')

with open(pubspec_path, 'r') as f:
    pubspec_content = f.readlines()

flutter_section_start_index = -1
assets_key_line_index = -1 # Index of the line containing 'assets:'

# 1. Find the top-level 'flutter:' section (not indented)
for i, line in enumerate(pubspec_content):
    if line.strip() == 'flutter:' and not line.startswith(' '): # Ensure it's a top-level key
        flutter_section_start_index = i
        break

if flutter_section_start_index == -1:
    print("Error: Top-level 'flutter:' section not found in pubspec.yaml. Manual intervention required.")
else:
    # 2. Search for 'assets:' within this top-level flutter section
    potential_assets_insert_point = flutter_section_start_index + 1

    for i in range(flutter_section_start_index + 1, len(pubspec_content)):
        line = pubspec_content[i]
        stripped_line = line.strip()

        # Break if we've moved past the 'flutter:' block (less indentation or new top-level key)
        if not line.startswith('  ') and stripped_line != '': # If line is not indented by 2 spaces and not empty
            break

        if stripped_line == 'assets:':
            assets_key_line_index = i
            break
        elif stripped_line == '# assets:':
            pubspec_content[i] = '  assets:\n' # Uncomment and normalize indentation
            assets_key_line_index = i
            break
        elif stripped_line == 'uses-material-design: true':
            potential_assets_insert_point = i + 1 # Suggest inserting after this common line

    # 3. If 'assets:' was not found, insert it at the determined point
    if assets_key_line_index == -1:
        # If no specific insertion point identified, find the end of the flutter block
        if potential_assets_insert_point == flutter_section_start_index + 1: # This means 'uses-material-design: true' was not found either
            # Search downwards from flutter_section_start_index for the last line indented by '  '
            temp_insert_point = flutter_section_start_index + 1
            for k in range(flutter_section_start_index + 1, len(pubspec_content)):
                if not pubspec_content[k].startswith('  ') and pubspec_content[k].strip() != '':
                    break # Found next top-level key
                temp_insert_point = k + 1
            potential_assets_insert_point = temp_insert_point


        # Ensure newline before inserting if the line above is not empty and doesn't end with newline
        if potential_assets_insert_point > 0 and not pubspec_content[potential_assets_insert_point-1].endswith('\n') and pubspec_content[potential_assets_insert_point-1].strip() != '':
             pubspec_content.insert(potential_assets_insert_point, '\n')
             potential_assets_insert_point += 1

        pubspec_content.insert(potential_assets_insert_point, '  assets:\n')
        assets_key_line_index = potential_assets_insert_point

    # 4. Add '    - assets/models/' under the 'assets:' key
    asset_to_add = '    - assets/models/\n'
    asset_already_declared = False

    # Check for existing 'assets/models/' declaration
    # Iterate from after assets_key_line_index until another major section or less indented line
    for i in range(assets_key_line_index + 1, len(pubspec_content)):
        line = pubspec_content[i]
        # Stop if we hit a line with less indentation than '    - ' or empty line after assets
        if line.strip() != '' and not line.startswith('    '):
            break
        if asset_to_add.strip() == line.strip(): # Compare stripped versions
            asset_already_declared = True
            break

    if not asset_already_declared:
        # Find the correct insertion point immediately after the last existing asset or the 'assets:' line itself
        insert_asset_line_index = assets_key_line_index + 1
        while insert_asset_line_index < len(pubspec_content) and \
              (pubspec_content[insert_asset_line_index].strip() == '' or \
               pubspec_content[insert_asset_line_index].startswith('    -')):
            insert_asset_line_index += 1

        pubspec_content.insert(insert_asset_line_index, asset_to_add)

        # Write the modified content back to pubspec.yaml
        with open(pubspec_path, 'w') as f:
            f.writelines(pubspec_content)
        print("Added 'assets/models/' to pubspec.yaml. Running flutter pub get...")
        os.chdir('ai_chatbot_app')
        !flutter pub get
        os.chdir('..')
        print("flutter pub get completed.")
    else:
        print("'assets/models/' already declared in pubspec.yaml. No changes made.")
````

````python
import os

pubspec_path = os.path.join('ai_chatbot_app', 'pubspec.yaml')

# Read the existing pubspec.yaml file
with open(pubspec_path, 'r') as f:
    pubspec_content = f.readlines()

updated_pubspec_content = []
flutter_section_found = False
assets_section_written = False
skip_old_assets = False

for i, line in enumerate(pubspec_content):
    stripped_line = line.strip()

    # Check for top-level 'flutter:' section
    if stripped_line == 'flutter:' and not line.startswith(' '):
        flutter_section_found = True
        updated_pubspec_content.append(line) # Add 'flutter:' line
        continue

    if flutter_section_found:
        # Stop processing flutter section if another top-level key is found
        if not line.startswith(' ') and stripped_line != '':
            flutter_section_found = False
            # If assets section wasn't written yet, write it before this new top-level section
            if not assets_section_written:
                updated_pubspec_content.append('  assets:\n')
                updated_pubspec_content.append('    - assets/models/\n')
                assets_section_written = True
            updated_pubspec_content.append(line)
            continue

        # Skip existing assets section (commented or uncommented) to write a clean one later
        if stripped_line == 'assets:' or stripped_line == '# assets:':
            skip_old_assets = True
            continue
        if skip_old_assets and (line.startswith('    -') or stripped_line == '' or stripped_line.startswith('#')): # Continue skipping children/comments
            if stripped_line != '' and not line.startswith('    -'): # Stop skipping if indent less than asset entry
                skip_old_assets = False
                updated_pubspec_content.append(line) # Add this line back
                continue
            continue # Keep skipping
        elif skip_old_assets:
            skip_old_assets = False # Stop skipping if not an asset line or its comment/blank

        if stripped_line == 'uses-material-design: true':
            updated_pubspec_content.append(line) # Add 'uses-material-design: true' line
            # Insert the new assets section immediately after uses-material-design: true
            if not assets_section_written:
                updated_pubspec_content.append('  assets:\n')
                updated_pubspec_content.append('    - assets/models/\n')
                assets_section_written = True
            continue

    # Add lines that are not part of flutter section or not skipped old assets
    if not skip_old_assets:
        updated_pubspec_content.append(line)

# If flutter section was found but assets weren't written (e.g., no uses-material-design: true)
# Append assets at the end of the flutter section or file if flutter not found
if flutter_section_found and not assets_section_written:
    # Find the last line that's part of the flutter block (indented by 2 spaces)
    last_flutter_line_index = -1
    for k in range(len(updated_pubspec_content) -1, -1, -1):
        if updated_pubspec_content[k].startswith('  '):
            last_flutter_line_index = k
            break
    if last_flutter_line_index != -1:
        # Insert after the last flutter-indented line
        updated_pubspec_content.insert(last_flutter_line_index + 1, '  assets:\n')
        updated_pubspec_content.insert(last_flutter_line_index + 2, '    - assets/models/\n')
    else:
        # Fallback: append at the very end (shouldn't happen with flutter: found)
        updated_pubspec_content.append('flutter:\n')
        updated_pubspec_content.append('  assets:\n')
        updated_pubspec_content.append('    - assets/models/\n')
    assets_section_written = True

# Check if 'assets/models/' was already present and prevent re-adding if it was
# This check is less critical now due to the reconstruction approach
# but as a final safeguard it ensures no double entries.
final_content_str = "".join(updated_pubspec_content)
if final_content_str.count('    - assets/models/\n') > 1:
    # If it somehow got added twice, remove the extra one
    # Simple hack: find the last occurrence and remove it
    first_occurrence = final_content_str.find('    - assets/models/\n')
    if first_occurrence != -1:
        second_occurrence = final_content_str.find('    - assets/models/\n', first_occurrence + len('    - assets/models/\n'))
        if second_occurrence != -1:
            final_content_str = final_content_str[:second_occurrence] + final_content_str[second_occurrence + len('    - assets/models/\n'):]
    updated_pubspec_content = final_content_str.splitlines(keepends=True)


# Write the modified content back to pubspec.yaml
with open(pubspec_path, 'w') as f:
    f.writelines(updated_pubspec_content)

print("Added 'assets/models/' to pubspec.yaml. Running flutter pub get...")
os.chdir('ai_chatbot_app')
!flutter pub get
os.chdir('..')
print("flutter pub get completed.")
````

This new code adopts a more robust strategy by performing two passes: first, it meticulously removes any incorrectly placed `assets:` and `- assets/models/` entries from the `dependencies` section; second, it ensures the correct `assets:` declaration and the `    - assets/models/` entry are properly inserted or maintained within the top-level `flutter:` section, ensuring correct YAML structure and functionality.



````python
import os

pubspec_path = os.path.join('ai_chatbot_app', 'pubspec.yaml')

# Read the existing pubspec.yaml file
with open(pubspec_path, 'r') as f:
    original_lines = f.readlines()

filtered_lines = []
in_dependencies_section = False

# --- Pass 1: Filter out problematic lines from the dependencies section ---
for i, line in enumerate(original_lines):
    stripped_line = line.strip()
    indentation = len(line) - len(line.lstrip())

    if stripped_line == 'dependencies:':
        in_dependencies_section = True
        filtered_lines.append(line)
        continue
    elif in_dependencies_section and (indentation < 2 and stripped_line != ''): # Exited dependencies block
        in_dependencies_section = False
        filtered_lines.append(line)
        continue
    elif stripped_line == 'dev_dependencies:': # Explicit end of dependencies
        in_dependencies_section = False
        filtered_lines.append(line)
        continue
    elif in_dependencies_section: # Inside dependencies block
        # Skip the incorrect assets from dependencies if they match the problematic indentation
        if (stripped_line == 'assets:' and indentation == 2) or \
           (stripped_line == '- assets/models/' and indentation == 4):
            continue
        filtered_lines.append(line)
        continue

    filtered_lines.append(line) # Keep all other lines outside of dependencies cleanup


# --- Pass 2: Ensure the correct assets structure is in the flutter section in the filtered_lines ---
final_lines = []
flutter_section_processed = False

for i, line in enumerate(filtered_lines):
    final_lines.append(line)

    # Detect the top-level 'flutter:' section once
    if line.strip() == 'flutter:' and not line.startswith(' ') and not flutter_section_processed:
        insert_point = i + 1 # Default insertion point right after 'flutter:'

        # Search for 'uses-material-design: true' within the flutter section in filtered_lines
        found_uses_material = False
        for temp_idx in range(i + 1, len(filtered_lines)):
            sub_line = filtered_lines[temp_idx]
            sub_stripped = sub_line.strip()
            sub_indentation = len(sub_line) - len(sub_line.lstrip())

            if sub_indentation < 2 and sub_stripped != '': # Exited flutter section
                break

            if sub_stripped == 'uses-material-design: true':
                insert_point = temp_idx + 1
                found_uses_material = True
                break

        # Check if 'assets:' and '    - assets/models/' are already correctly present
        assets_block_already_present_correctly = False
        models_entry_already_present_correctly = False

        current_search_idx = i + 1 # Start search from after 'flutter:'
        while current_search_idx < len(filtered_lines):
            search_line = filtered_lines[current_search_idx]
            search_stripped = search_line.strip()
            search_indentation = len(search_line) - len(search_line.lstrip())

            if search_indentation < 2 and search_stripped != '': # Exited flutter block
                break

            if search_stripped == 'assets:' and search_indentation == 2:
                assets_block_already_present_correctly = True
                # Check for models entry right after it or within its sub-block
                sub_models_idx = current_search_idx + 1
                while sub_models_idx < len(filtered_lines):
                    models_line = filtered_lines[sub_models_idx]
                    models_stripped = models_line.strip()
                    models_indentation = len(models_line) - len(models_line.lstrip())
                    if models_indentation < 4 and models_stripped != '': # Exited assets sub-block
                        break
                    if models_stripped == '- assets/models/' and models_indentation == 4:
                        models_entry_already_present_correctly = True
                        break
                    sub_models_idx += 1
                break # Found assets block
            current_search_idx += 1


        # If the assets block is not correctly present, insert it
        if not assets_block_already_present_correctly:
            final_lines.insert(insert_point, '  assets:\n')
            final_lines.insert(insert_point + 1, '    - assets/models/\n')
        # If assets block is present but the models entry is missing, insert just the models entry
        elif assets_block_already_present_correctly and not models_entry_already_present_correctly:
            # Find the exact index of the '  assets:' line in final_lines to insert after it
            actual_assets_line_idx = -1
            for k, l in enumerate(final_lines):
                if l.strip() == 'assets:' and l.startswith('  '): # Correct indentation for flutter assets
                    # Ensure it's truly within the flutter block by checking parentage (simple heuristic)
                    temp_j = k - 1
                    while temp_j >= 0 and final_lines[temp_j].startswith(' '): temp_j -= 1
                    if temp_j >= 0 and final_lines[temp_j].strip() == 'flutter:':
                        actual_assets_line_idx = k
                        break

            if actual_assets_line_idx != -1:
                insert_point_for_models = actual_assets_line_idx + 1
                while insert_point_for_models < len(final_lines) and \
                      (final_lines[insert_point_for_models].strip() == '' or \
                       final_lines[insert_point_for_models].startswith('    -')):
                    insert_point_for_models += 1
                final_lines.insert(insert_point_for_models, '    - assets/models/\n')

        flutter_section_processed = True # Mark flutter section as processed


# Write the modified content back to pubspec.yaml
final_content = "".join(final_lines)
with open(pubspec_path, 'w') as f:
    f.write(final_content)

print("Attempting to fix pubspec.yaml and running flutter pub get...")
os.chdir('ai_chatbot_app')
!flutter pub get
os.chdir('..')
print("flutter pub get completed.")
````

**Reasoning**:
The next step is to create the `chatbot_provider.dart` file as per instruction 2. This provider will manage the state of the chatbot UI and interact with the `ChatbotService`.



````python
import os

file_path = os.path.join('ai_chatbot_app', 'lib', 'presentation', 'providers', 'chatbot_provider.dart')

file_content = '''
import 'package:flutter/material.dart';
import 'package:ai_chatbot_app/domain/usecases/chatbot_service.dart';

class ChatMessage {
  final String text;
  final bool isUser;

  ChatMessage({required this.text, required this.isUser});
}

class ChatbotProvider with ChangeNotifier {
  final ChatbotService _chatbotService;
  final List<ChatMessage> _messages = [];
  bool _isLoading = false;

  ChatbotProvider(this._chatbotService);

  List<ChatMessage> get messages => _messages;
  bool get isLoading => _isLoading;

  Future<void> sendMessage(String text) async {
    if (text.trim().isEmpty) return;

    _messages.add(ChatMessage(text: text, isUser: true));
    _isLoading = true;
    notifyListeners();

    try {
      final response = await _chatbotService.sendMessage(text);
      _messages.add(ChatMessage(text: response, isUser: false));
    } catch (e) {
      _messages.add(ChatMessage(text: 'Error: Could not get a response. ($e)', isUser: false));
      print('ChatbotProvider error: $e');
    }

    _isLoading = false;
    notifyListeners();
  }
}
'''

with open(file_path, 'w') as f:
    f.write(file_content)

print(f"Created ChatbotProvider at {file_path}")
````

**Reasoning**:
The next step is to create the `chatbot_page.dart` file as per instruction 3. This page will define the visual layout and user interaction for the chatbot.



````python
import os

file_path = os.path.join('ai_chatbot_app', 'lib', 'presentation', 'pages', 'chatbot_page.dart')

file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/presentation/providers/chatbot_provider.dart';

class ChatbotPage extends StatefulWidget {
  const ChatbotPage({super.key});

  @override
  State<ChatbotPage> createState() => _ChatbotPageState();
}

class _ChatbotPageState extends State<ChatbotPage> {
  final TextEditingController _textController = TextEditingController();
  final ScrollController _scrollController = ScrollController();

  @override
  void dispose() {
    _textController.dispose();
    _scrollController.dispose();
    super.dispose();
  }

  void _sendMessage() {
    if (_textController.text.isNotEmpty) {
      Provider.of<ChatbotProvider>(context, listen: false)
          .sendMessage(_textController.text);
      _textController.clear();
      _scrollToBottom();
    }
  }

  void _scrollToBottom() {
    // Ensure the scroll happens after the new message is rendered
    WidgetsBinding.instance.addPostFrameCallback((_) {
      _scrollController.animateTo(
        _scrollController.position.maxScrollExtent,
        duration: const Duration(milliseconds: 300),
        curve: Curves.easeOut,
      );
    });
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        title: const Text('AI Chatbot'),
      ),
      body: Column(
        children: <Widget>[
          Expanded(
            child: Consumer<ChatbotProvider>(
              builder: (context, chatbotProvider, child) {
                // Scroll to bottom whenever messages change
                _scrollToBottom();
                return ListView.builder(
                  controller: _scrollController,
                  itemCount: chatbotProvider.messages.length,
                  itemBuilder: (context, index) {
                    final message = chatbotProvider.messages[index];
                    return Align(
                      alignment: message.isUser
                          ? Alignment.centerRight
                          : Alignment.centerLeft,
                      child: Container(
                        margin: const EdgeInsets.symmetric(
                            vertical: 5.0, horizontal: 10.0),
                        padding: const EdgeInsets.all(10.0),
                        decoration: BoxDecoration(
                          color: message.isUser ? Colors.blue[100] : Colors.grey[300],
                          borderRadius: BorderRadius.circular(15.0),
                        ),
                        child: Text(message.text),
                      ),
                    );
                  },
                );
              },
            ),
          ),
          if (Provider.of<ChatbotProvider>(context).isLoading)
            const Padding(
              padding: EdgeInsets.all(8.0),
              child: LinearProgressIndicator(),
            ),
          Padding(
            padding: const EdgeInsets.all(8.0),
            child: Row(
              children: <Widget>[
                Expanded(
                  child: TextField(
                    controller: _textController,
                    decoration: InputDecoration(
                      hintText: 'Enter your message...',
                      border: OutlineInputBorder(borderRadius: BorderRadius.circular(20.0)),
                      contentPadding: const EdgeInsets.symmetric(horizontal: 20.0, vertical: 10.0),
                    ),
                    onSubmitted: (_) => _sendMessage(),
                  ),
                ),
                const SizedBox(width: 8.0),
                FloatingActionButton(
                  onPressed: _sendMessage,
                  child: const Icon(Icons.send),
                ),
              ],
            ),
          ),
        ],
      ),
    );
  }
}
'''

with open(file_path, 'w') as f:
    f.write(file_content)

print(f
````

````python
import os

file_path = os.path.join('ai_chatbot_app', 'lib', 'presentation', 'pages', 'chatbot_page.dart')

file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/presentation/providers/chatbot_provider.dart';

class ChatbotPage extends StatefulWidget {
  const ChatbotPage({super.key});

  @override
  State<ChatbotPage> createState() => _ChatbotPageState();
}

class _ChatbotPageState extends State<ChatbotPage> {
  final TextEditingController _textController = TextEditingController();
  final ScrollController _scrollController = ScrollController();

  @override
  void dispose() {
    _textController.dispose();
    _scrollController.dispose();
    super.dispose();
  }

  void _sendMessage() {
    if (_textController.text.isNotEmpty) {
      Provider.of<ChatbotProvider>(context, listen: false)
          .sendMessage(_textController.text);
      _textController.clear();
      _scrollToBottom();
    }
  }

  void _scrollToBottom() {
    // Ensure the scroll happens after the new message is rendered
    WidgetsBinding.instance.addPostFrameCallback((_) {
      _scrollController.animateTo(
        _scrollController.position.maxScrollExtent,
        duration: const Duration(milliseconds: 300),
        curve: Curves.easeOut,
      );
    });
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        title: const Text('AI Chatbot'),
      ),
      body: Column(
        children: <Widget>[
          Expanded(
            child: Consumer<ChatbotProvider>(
              builder: (context, chatbotProvider, child) {
                // Scroll to bottom whenever messages change
                _scrollToBottom();
                return ListView.builder(
                  controller: _scrollController,
                  itemCount: chatbotProvider.messages.length,
                  itemBuilder: (context, index) {
                    final message = chatbotProvider.messages[index];
                    return Align(
                      alignment: message.isUser
                          ? Alignment.centerRight
                          : Alignment.centerLeft,
                      child: Container(
                        margin: const EdgeInsets.symmetric(
                            vertical: 5.0, horizontal: 10.0),
                        padding: const EdgeInsets.all(10.0),
                        decoration: BoxDecoration(
                          color: message.isUser ? Colors.blue[100] : Colors.grey[300],
                          borderRadius: BorderRadius.circular(15.0),
                        ),
                        child: Text(message.text),
                      ),
                    );
                  },
                );
              },
            ),
          ),
          if (Provider.of<ChatbotProvider>(context).isLoading)
            const Padding(
              padding: EdgeInsets.all(8.0),
              child: LinearProgressIndicator(),
            ),
          Padding(
            padding: const EdgeInsets.all(8.0),
            child: Row(
              children: <Widget>[
                Expanded(
                  child: TextField(
                    controller: _textController,
                    decoration: InputDecoration(
                      hintText: 'Enter your message...',
                      border: OutlineInputBorder(borderRadius: BorderRadius.circular(20.0)),
                      contentPadding: const EdgeInsets.symmetric(horizontal: 20.0, vertical: 10.0),
                    ),
                    onSubmitted: (_) => _sendMessage(),
                  ),
                ),
                const SizedBox(width: 8.0),
                FloatingActionButton(
                  onPressed: _sendMessage,
                  child: const Icon(Icons.send),
                ),
              ],
            ),
          ),
        ],
      ),
    );
  }
}
'''

with open(file_path, 'w') as f:
    f.write(file_content)

print(f"Created ChatbotPage at {file_path}")
````

**Reasoning**:
The first instruction is to create a new Dart file named `image_generation_service.dart` in `ai_chatbot_app/lib/domain/usecases`. This file will define an abstract class `ImageGenerationService` and its concrete implementation `ImageGenerationServiceImpl` that interacts with the `AiDataSource`.



````python
import os

file_path = os.path.join('ai_chatbot_app', 'lib', 'domain', 'usecases', 'image_generation_service.dart')

file_content = '''
import 'package:ai_chatbot_app/data/datasources/ai_data_source.dart';

// Abstract interface for text-to-image generation
abstract class ImageGenerationService {
  Future<List<int>> generateImage(String prompt);
}

// Concrete implementation of ImageGenerationService
class ImageGenerationServiceImpl implements ImageGenerationService {
  final AiDataSource _aiDataSource;

  ImageGenerationServiceImpl(this._aiDataSource);

  @override
  Future<List<int>> generateImage(String prompt) async {
    // Here, you might add business logic before calling the data source
    // For example, input validation, prompt refinement, etc.
    return await _aiDataSource.generateImageFromText(prompt);
  }
}
'''

with open(file_path, 'w') as f:
    f.write(file_content)

print(f"Created ImageGenerationService at {file_path}")
````

**Reasoning**:
The next step is to create the `image_generation_provider.dart` file as per instruction 2. This provider will manage the state for text-to-image generation, including the text prompt, the generated image data (`List<int>`), and a loading state.



````python
import os

file_path = os.path.join('ai_chatbot_app', 'lib', 'presentation', 'providers', 'image_generation_provider.dart')

file_content = '''
import 'package:flutter/material.dart';
import 'package:ai_chatbot_app/domain/usecases/image_generation_service.dart';

class ImageGenerationProvider with ChangeNotifier {
  final ImageGenerationService _imageGenerationService;
  List<int>? _generatedImageBytes;
  bool _isLoading = false;
  String? _errorMessage;

  ImageGenerationProvider(this._imageGenerationService);

  List<int>? get generatedImageBytes => _generatedImageBytes;
  bool get isLoading => _isLoading;
  String? get errorMessage => _errorMessage;

  Future<void> generateImage(String prompt) async {
    if (prompt.trim().isEmpty) {
      _errorMessage = 'Please enter a prompt to generate an image.';
      notifyListeners();
      return;
    }

    _isLoading = true;
    _errorMessage = null;
    _generatedImageBytes = null; // Clear previous image
    notifyListeners();

    try {
      _generatedImageBytes = await _imageGenerationService.generateImage(prompt);
    } catch (e) {
      _errorMessage = 'Error generating image: $e';
      print('ImageGenerationProvider error: $e');
    }

    _isLoading = false;
    notifyListeners();
  }

  void clearImage() {
    _generatedImageBytes = null;
    _errorMessage = null;
    notifyListeners();
  }
}
'''

with open(file_path, 'w') as f:
    f.write(file_content)

print(f"Created ImageGenerationProvider at {file_path}")
````

## Image Generation Feature Development (Uploaded Image + Prompt)

### Subtask:
Implement the functionality to allow users to upload an image and provide a text prompt to generate a modified or new image. This requires integrating with image picker functionalities and an appropriate image-to-image AI model.


**Reasoning**:
The first instruction is to create a new Dart file named `image_modification_service.dart` in `ai_chatbot_app/lib/domain/usecases`. This file will define an abstract class `ImageModificationService` and its concrete implementation `ImageModificationServiceImpl` that interacts with the `AiDataSource` to perform image modification.



````python
import os

file_path = os.path.join('ai_chatbot_app', 'lib', 'domain', 'usecases', 'image_modification_service.dart')

file_content = '''
import 'package:ai_chatbot_app/data/datasources/ai_data_source.dart';

// Abstract interface for image modification
abstract class ImageModificationService {
  Future<List<int>> modifyImage(List<int> imageData, String prompt);
}

// Concrete implementation of ImageModificationService
class ImageModificationServiceImpl implements ImageModificationService {
  final AiDataSource _aiDataSource;

  ImageModificationServiceImpl(this._aiDataSource);

  @override
  Future<List<int>> modifyImage(List<int> imageData, String prompt) async {
    // Here, you might add business logic before calling the data source
    // For example, input validation, prompt refinement, etc.
    return await _aiDataSource.modifyImage(imageData, prompt);
  }
}
'''

with open(file_path, 'w') as f:
    f.write(file_content)

print(f"Created ImageModificationService at {file_path}")
````

**Reasoning**:
The next step is to create the `image_modification_provider.dart` file as per instruction 2. This provider will manage the state for image modification, including the original image bytes (`List<int>?`), the text prompt (`String`), the modified image bytes (`List<int>?`), a loading state (`bool`), and any error messages (`String?`). It should have methods to pick an image, set a prompt, and call the `ImageModificationService` to modify the image.



````python
import os

file_path = os.path.join('ai_chatbot_app', 'lib', 'presentation', 'providers', 'image_modification_provider.dart')

file_content = '''
import 'package:flutter/material.dart';
import 'package:image_picker/image_picker.dart';
import 'dart:io';
import 'package:ai_chatbot_app/domain/usecases/image_modification_service.dart';

class ImageModificationProvider with ChangeNotifier {
  final ImageModificationService _imageModificationService;

  File? _originalImageFile;
  List<int>? _originalImageBytes;
  String _prompt = '';
  List<int>? _modifiedImageBytes;
  bool _isLoading = false;
  String? _errorMessage;

  ImageModificationProvider(this._imageModificationService);

  File? get originalImageFile => _originalImageFile;
  List<int>? get originalImageBytes => _originalImageBytes;
  String get prompt => _prompt;
  List<int>? get modifiedImageBytes => _modifiedImageBytes;
  bool get isLoading => _isLoading;
  String? get errorMessage => _errorMessage;

  void setPrompt(String newPrompt) {
    _prompt = newPrompt;
    notifyListeners();
  }

  Future<void> pickImage() async {
    final ImagePicker picker = ImagePicker();
    final XFile? image = await picker.pickImage(source: ImageSource.gallery);

    if (image != null) {
      _originalImageFile = File(image.path);
      _originalImageBytes = await _originalImageFile!.readAsBytes();
      _modifiedImageBytes = null; // Clear any previous modified image
      _errorMessage = null;
      notifyListeners();
    }
  }

  Future<void> modifyImage() async {
    if (_originalImageBytes == null) {
      _errorMessage = 'Please pick an image first.';
      notifyListeners();
      return;
    }
    if (_prompt.trim().isEmpty) {
      _errorMessage = 'Please enter a prompt for modification.';
      notifyListeners();
      return;
    }

    _isLoading = true;
    _errorMessage = null;
    _modifiedImageBytes = null; // Clear previous modified image
    notifyListeners();

    try {
      _modifiedImageBytes = await _imageModificationService.modifyImage(
          _originalImageBytes!, _prompt);
    } catch (e) {
      _errorMessage = 'Error modifying image: $e';
      print('ImageModificationProvider error: $e');
    }

    _isLoading = false;
    notifyListeners();
  }

  void clearAll() {
    _originalImageFile = null;
    _originalImageBytes = null;
    _prompt = '';
    _modifiedImageBytes = null;
    _errorMessage = null;
    notifyListeners();
  }
}
'''

with open(file_path, 'w') as f:
    f.write(file_content)

print(f"Created ImageModificationProvider at {file_path}")
````

**Reasoning**:
The next step is to create the `image_modification_page.dart` file as per instruction 3. This page will provide the user interface for image modification, including controls for picking an image, entering a prompt, triggering modification, and displaying results.



````python
import os

file_path = os.path.join('ai_chatbot_app', 'lib', 'presentation', 'pages', 'image_modification_page.dart')

file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/presentation/providers/image_modification_provider.dart';
import 'dart:typed_data'; // Required for Uint8List.fromList

class ImageModificationPage extends StatefulWidget {
  const ImageModificationPage({super.key});

  @override
  State<ImageModificationPage> createState() => _ImageModificationPageState();
}

class _ImageModificationPageState extends State<ImageModificationPage> {
  final TextEditingController _promptController = TextEditingController();

  @override
  void initState() {
    super.initState();
    // Listen for prompt changes
    _promptController.addListener(() {
      Provider.of<ImageModificationProvider>(context, listen: false)
          .setPrompt(_promptController.text);
    });
  }

  @override
  void dispose() {
    _promptController.dispose();
    super.dispose();
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        title: const Text('Image Modifier'),
        actions: [
          IconButton(
            icon: const Icon(Icons.clear_all),
            onPressed: () {
              Provider.of<ImageModificationProvider>(context, listen: false).clearAll();
              _promptController.clear();
            },
          ),
        ],
      ),
      body: Consumer<ImageModificationProvider>(
        builder: (context, imageModifierProvider, child) {
          return SingleChildScrollView(
            padding: const EdgeInsets.all(16.0),
            child: Column(
              crossAxisAlignment: CrossAxisAlignment.stretch,
              children: <Widget>[
                // Image Picker Button
                ElevatedButton.icon(
                  onPressed: imageModifierProvider.pickImage,
                  icon: const Icon(Icons.image),
                  label: const Text('Pick Image'),
                ),
                const SizedBox(height: 16.0),

                // Original Image Display
                if (imageModifierProvider.originalImageBytes != null)
                  Column(
                    children: [
                      const Text('Original Image:', style: TextStyle(fontWeight: FontWeight.bold)),
                      const SizedBox(height: 8.0),
                      Image.memory(Uint8List.fromList(imageModifierProvider.originalImageBytes!),
                          height: 200, fit: BoxFit.contain),
                      const SizedBox(height: 16.0),
                    ],
                  ),

                // Prompt Input
                TextField(
                  controller: _promptController,
                  decoration: InputDecoration(
                    hintText: 'Enter modification prompt (e.g., "make it vintage", "add sunglasses")',
                    border: OutlineInputBorder(borderRadius: BorderRadius.circular(10.0)),
                    contentPadding: const EdgeInsets.symmetric(horizontal: 12.0, vertical: 8.0),
                  ),
                  maxLines: 3,
                ),
                const SizedBox(height: 16.0),

                // Modify Button
                imageModifierProvider.isLoading
                    ? const Center(child: CircularProgressIndicator())
                    : ElevatedButton.icon(
                        onPressed: imageModifierProvider.originalImageBytes != null && _promptController.text.isNotEmpty
                            ? imageModifierProvider.modifyImage
                            : null,
                        icon: const Icon(Icons.auto_fix_high),
                        label: const Text('Modify Image'),
                      ),
                const SizedBox(height: 16.0),

                // Error Message
                if (imageModifierProvider.errorMessage != null)
                  Text(
                    imageModifierProvider.errorMessage!,
                    style: const TextStyle(color: Colors.red),
                    textAlign: TextAlign.center,
                  ),
                const SizedBox(height: 16.0),

                // Modified Image Display
                if (imageModifierProvider.modifiedImageBytes != null)
                  Column(
                    children: [
                      const Text('Modified Image:', style: TextStyle(fontWeight: FontWeight.bold)),
                      const SizedBox(height: 8.0),
                      Image.memory(Uint8List.fromList(imageModifierProvider.modifiedImageBytes!),
                          height: 200, fit: BoxFit.contain),
                      const SizedBox(height: 16.0),
                    ],
                  ),
              ],
            ),
          );
        },
      ),
    );
  }
}
'''

with open(file_path, 'w') as f:
    f.write(file_content)

print(f"Created ImageModificationPage at {file_path}")
````

````python
import os

file_path = os.path.join('ai_chatbot_app', 'lib', 'presentation', 'pages', 'chatbot_page.dart')

file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider';
import 'package:ai_chatbot_app/presentation/providers/chatbot_provider.dart';

class ChatbotPage extends StatefulWidget {
  const ChatbotPage({super.key});

  @override
  State<ChatbotPage> createState() => _ChatbotPageState();
}

class _ChatbotPageState extends State<ChatbotPage> {
  final TextEditingController _textController = TextEditingController();
  final ScrollController _scrollController = ScrollController();

  @override
  void dispose() {
    _textController.dispose();
    _scrollController.dispose();
    super.dispose();
  }

  void _sendMessage() {
    if (_textController.text.isNotEmpty) {
      Provider.of<ChatbotProvider>(context, listen: false)
          .sendMessage(_textController.text);
      _textController.clear();
      // _scrollToBottom(); // Auto-scroll is now handled by the Consumer's builder
    }
  }

  void _scrollToBottom() {
    // Ensure the scroll happens after the new message is rendered
    WidgetsBinding.instance.addPostFrameCallback((_) {
      if (_scrollController.hasClients) {
        _scrollController.animateTo(
          _scrollController.position.maxScrollExtent,
          duration: const Duration(milliseconds: 300),
          curve: Curves.easeOut,
        );
      }
    });
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        title: const Text('AI Chatbot'),
      ),
      body: Column(
        children: <Widget>[
          Expanded(
            child: Consumer<ChatbotProvider>(
              builder: (context, chatbotProvider, child) {
                // Scroll to bottom whenever messages change
                if (chatbotProvider.messages.isNotEmpty && _scrollController.hasClients) {
                   _scrollToBottom();
                }
                return ListView.builder(
                  controller: _scrollController,
                  padding: const EdgeInsets.symmetric(horizontal: 10.0, vertical: 8.0),
                  itemCount: chatbotProvider.messages.length,
                  itemBuilder: (context, index) {
                    final message = chatbotProvider.messages[index];
                    final isUser = message.isUser;
                    return Padding(
                      padding: const EdgeInsets.symmetric(vertical: 4.0),
                      child: Row(
                        mainAxisAlignment: isUser
                            ? MainAxisAlignment.end
                            : MainAxisAlignment.start,
                        crossAxisAlignment: CrossAxisAlignment.end,
                        children: <Widget>[
                          if (!isUser) ...[
                            CircleAvatar(
                              backgroundColor: Theme.of(context).primaryColor,
                              child: const Icon(Icons.robot, color: Colors.white, size: 20),
                            ),
                            const SizedBox(width: 8.0),
                          ],
                          Flexible(
                            child: Container(
                              constraints: BoxConstraints(
                                maxWidth: MediaQuery.of(context).size.width * 0.75,
                              ),
                              padding: const EdgeInsets.all(12.0),
                              decoration: BoxDecoration(
                                color: isUser
                                    ? Theme.of(context).colorScheme.primary
                                    : Theme.of(context).colorScheme.secondary.withOpacity(0.2),
                                borderRadius: BorderRadius.only(
                                  topLeft: const Radius.circular(15.0),
                                  topRight: const Radius.circular(15.0),
                                  bottomLeft: isUser
                                      ? const Radius.circular(15.0)
                                      : const Radius.circular(5.0),
                                  bottomRight: isUser
                                      ? const Radius.circular(5.0)
                                      : const Radius.circular(15.0),
                                ),
                              ),
                              child: Text(
                                message.text,
                                style: TextStyle(
                                  color: isUser ? Colors.white : Theme.of(context).textTheme.bodyLarge?.color,
                                ),
                              ),
                            ),
                          ),
                          if (isUser) ...[
                            const SizedBox(width: 8.0),
                            CircleAvatar(
                              backgroundColor: Theme.of(context).primaryColor,
                              child: const Icon(Icons.person, color: Colors.white, size: 20),
                            ),
                          ],
                        ],
                      ),
                    );
                  },
                );
              },
            ),
          ),
          if (Provider.of<ChatbotProvider>(context).isLoading)
            const Padding(
              padding: EdgeInsets.all(8.0),
              child: LinearProgressIndicator(),
            ),
          Padding(
            padding: const EdgeInsets.all(8.0),
            child: Row(
              children: <Widget>[
                Expanded(
                  child: TextField(
                    controller: _textController,
                    decoration: Theme.of(context).inputDecorationTheme.copyWith(
                      hintText: 'Enter your message...',
                    ),
                    onSubmitted: (_) => _sendMessage(),
                  ),
                ),
                const SizedBox(width: 8.0),
                FloatingActionButton(
                  onPressed: _sendMessage,
                  backgroundColor: Theme.of(context).colorScheme.secondary,
                  foregroundColor: Theme.of(context).colorScheme.onSecondary,
                  child: const Icon(Icons.send),
                ),
              ],
            ),
          ),
        ],
      ),
    );
  }
}
'''

with open(file_path, 'w') as f:
    f.write(file_content)

print(f
````

````python
import os

file_path = os.path.join('ai_chatbot_app', 'lib', 'presentation', 'pages', 'image_generation_page.dart')

file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider';
import 'package:ai_chatbot_app/presentation/providers/image_generation_provider.dart';
import 'dart:typed_data'; // Required for Uint8List.fromList

class ImageGenerationPage extends StatefulWidget {
  const ImageGenerationPage({super.key});

  @override
  State<ImageGenerationPage> createState() => _ImageGenerationPageState();
}

class _ImageGenerationPageState extends State<ImageGenerationPage> {
  final TextEditingController _promptController = TextEditingController();

  @override
  void dispose() {
    _promptController.dispose();
    super.dispose();
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        title: const Text('Text-to-Image Generator'),
      ),
      body: Consumer<ImageGenerationProvider>(
        builder: (context, imageProvider, child) {
          return SingleChildScrollView(
            padding: const EdgeInsets.all(16.0),
            child: Column(
              crossAxisAlignment: CrossAxisAlignment.stretch,
              children: <Widget>[
                // Prompt Input
                TextField(
                  controller: _promptController,
                  decoration: Theme.of(context).inputDecorationTheme.copyWith(
                    hintText: 'Enter your image prompt (e.g., "a cat in space")',
                  ),
                  maxLines: 3,
                ),
                const SizedBox(height: 16.0),

                // Generate Button
                imageProvider.isLoading
                    ? Center(
                        child: Column(
                          children: [
                            const CircularProgressIndicator(),
                            const SizedBox(height: 8.0),
                            Text('Generating image...', style: Theme.of(context).textTheme.bodyMedium),
                          ],
                        ),
                      )
                    : ElevatedButton(
                        onPressed: () {
                          imageProvider.generateImage(_promptController.text);
                        },
                        child: const Text('Generate Image'),
                      ),
                const SizedBox(height: 16.0),

                // Error Message
                if (imageProvider.errorMessage != null)
                  Padding(
                    padding: const EdgeInsets.only(bottom: 16.0),
                    child: Text(
                      imageProvider.errorMessage!,
                      style: Theme.of(context).textTheme.bodyMedium?.copyWith(color: Colors.red),
                      textAlign: TextAlign.center,
                    ),
                  ),

                // Generated Image Display
                if (imageProvider.generatedImageBytes != null)
                  Column(
                    children: [
                      const Text('Generated Image:', style: TextStyle(fontWeight: FontWeight.bold)),
                      const SizedBox(height: 8.0),
                      Container(
                        decoration: BoxDecoration(
                          border: Border.all(color: Theme.of(context).colorScheme.primary.withOpacity(0.5), width: 2),
                          borderRadius: BorderRadius.circular(10),
                        ),
                        child: ClipRRect(
                          borderRadius: BorderRadius.circular(8),
                          child: Image.memory(
                            Uint8List.fromList(imageProvider.generatedImageBytes!),
                            fit: BoxFit.contain,
                            errorBuilder: (context, error, stackTrace) => const Icon(Icons.error_outline, color: Colors.red, size: 50),
                          ),
                        ),
                      ),
                      const SizedBox(height: 16.0),
                      ElevatedButton.icon(
                        onPressed: imageProvider.clearImage,
                        icon: const Icon(Icons.clear),
                        label: const Text('Clear Image'),
                      ),
                    ],
                  )
                else if (!imageProvider.isLoading) // Show placeholder only when not loading and no image
                  Padding(
                    padding: const EdgeInsets.all(24.0),
                    child: Text(
                      'No image generated yet.\nEnter a prompt and click "Generate Image".',
                      textAlign: TextAlign.center,
                      style: Theme.of(context).textTheme.bodyMedium?.copyWith(color: Colors.grey[600]),
                    ),
                  ),
              ],
            ),
          );
        },
      ),
    );
  }
}
'''

with open(file_path, 'w') as f:
    f.write(file_content)

print(f"Updated ImageGenerationPage at {file_path} with improved UI and visual cues.")
````

````python
import os

file_path = os.path.join('ai_chatbot_app', 'lib', 'presentation', 'pages', 'image_modification_page.dart')

file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider';
import 'package:ai_chatbot_app/presentation/providers/image_modification_provider.dart';
import 'dart:typed_data'; // Required for Uint8List.fromList

class ImageModificationPage extends StatefulWidget {
  const ImageModificationPage({super.key});

  @override
  State<ImageModificationPage> createState() => _ImageModificationPageState();
}

class _ImageModificationPageState extends State<ImageModificationPage> {
  final TextEditingController _promptController = TextEditingController();

  @override
  void initState() {
    super.initState();
    // Listen for prompt changes
    _promptController.addListener(() {
      Provider.of<ImageModificationProvider>(context, listen: false)
          .setPrompt(_promptController.text);
    });
  }

  @override
  void dispose() {
    _promptController.dispose();
    super.dispose();
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        title: const Text('Image Modifier'),
        actions: [
          IconButton(
            icon: const Icon(Icons.clear_all),
            onPressed: () {
              Provider.of<ImageModificationProvider>(context, listen: false).clearAll();
              _promptController.clear();
            },
          ),
        ],
      ),
      body: Consumer<ImageModificationProvider>(
        builder: (context, imageModifierProvider, child) {
          return SingleChildScrollView(
            padding: const EdgeInsets.all(16.0),
            child: Column(
              crossAxisAlignment: CrossAxisAlignment.stretch,
              children: <Widget>[
                // Image Picker Button
                ElevatedButton.icon(
                  onPressed: imageModifierProvider.pickImage,
                  icon: const Icon(Icons.image),
                  label: const Text('Pick Image to Modify'),
                  style: Theme.of(context).elevatedButtonTheme.style?.copyWith(
                    backgroundColor: MaterialStateProperty.all(Theme.of(context).primaryColor),
                    foregroundColor: MaterialStateProperty.all(Colors.white),
                  ),
                ),
                const SizedBox(height: 16.0),

                // Prompt Input
                TextField(
                  controller: _promptController,
                  decoration: Theme.of(context).inputDecorationTheme.copyWith(
                    hintText: 'Enter modification prompt (e.g., "make it vintage", "add sunglasses")',
                  ),
                  maxLines: 3,
                ),
                const SizedBox(height: 16.0),

                // Modify Button
                imageModifierProvider.isLoading
                    ? Center(
                        child: Column(
                          children: [
                            const CircularProgressIndicator(),
                            const SizedBox(height: 8.0),
                            Text('Modifying image...', style: Theme.of(context).textTheme.bodyMedium),
                          ],
                        ),
                      )
                    : ElevatedButton.icon(
                        onPressed: imageModifierProvider.originalImageBytes != null && _promptController.text.isNotEmpty
                            ? imageModifierProvider.modifyImage
                            : null,
                        icon: const Icon(Icons.auto_fix_high),
                        label: const Text('Apply Modification'),
                      ),
                const SizedBox(height: 16.0),

                // Error Message
                if (imageModifierProvider.errorMessage != null)
                  Padding(
                    padding: const EdgeInsets.only(bottom: 16.0),
                    child: Text(
                      imageModifierProvider.errorMessage!,
                      style: Theme.of(context).textTheme.bodyMedium?.copyWith(color: Colors.red),
                      textAlign: TextAlign.center,
                    ),
                  ),

                // Original Image Display
                if (imageModifierProvider.originalImageBytes != null)
                  Column(
                    crossAxisAlignment: CrossAxisAlignment.stretch,
                    children: [
                      Text('Original Image:', style: Theme.of(context).textTheme.headlineSmall?.copyWith(fontWeight: FontWeight.bold)),
                      const SizedBox(height: 8.0),
                      Container(
                        decoration: BoxDecoration(
                          border: Border.all(color: Theme.of(context).colorScheme.primary.withOpacity(0.5), width: 2),
                          borderRadius: BorderRadius.circular(10),
                        ),
                        child: ClipRRect(
                          borderRadius: BorderRadius.circular(8),
                          child: Image.memory(Uint8List.fromList(imageModifierProvider.originalImageBytes!),
                              height: 200, fit: BoxFit.contain),
                        ),
                      ),
                      const SizedBox(height: 16.0),
                    ],
                  ),

                // Modified Image Display
                if (imageModifierProvider.modifiedImageBytes != null)
                  Column(
                    crossAxisAlignment: CrossAxisAlignment.stretch,
                    children: [
                      Text('Modified Image:', style: Theme.of(context).textTheme.headlineSmall?.copyWith(fontWeight: FontWeight.bold)),
                      const SizedBox(height: 8.0),
                      Container(
                        decoration: BoxDecoration(
                          border: Border.all(color: Theme.of(context).colorScheme.secondary.withOpacity(0.5), width: 2),
                          borderRadius: BorderRadius.circular(10),
                        ),
                        child: ClipRRect(
                          borderRadius: BorderRadius.circular(8),
                          child: Image.memory(Uint8List.fromList(imageModifierProvider.modifiedImageBytes!),
                              height: 200, fit: BoxFit.contain),
                        ),
                      ),
                      const SizedBox(height: 16.0),
                    ],
                  ),
              ],
            ),
          );
        },
      ),
    );
  }
}
'''

with open(file_path, 'w') as f:
    f.write(file_content)

print(f
````

### 6. Ensure Interactive Elements Have Visual Feedback

The current implementation largely addresses the need for appropriate visual feedback for interactive elements through a combination of global theme settings and specific widget implementations:

*   **Buttons (`ElevatedButton`, `FloatingActionButton`)**: The `ThemeData` defined in `main.dart` includes `elevatedButtonTheme`, which sets a consistent style for `ElevatedButton` across the app. When `onPressed` is `null`, Flutter automatically styles the button for a disabled state, providing visual feedback. `FloatingActionButton` also handles its states inherently. Loading indicators (e.g., `CircularProgressIndicator` or `LinearProgressIndicator`) are explicitly shown when an action is in progress (e.g., `imageProvider.isLoading` or `chatbotProvider.isLoading`).

*   **Text Fields (`TextField`)**: The `inputDecorationTheme` in `main.dart` provides a unified look for text fields. Flutter's Material Design inherently provides visual cues for focus (e.g., accent color border) and input states. Disabled states for text fields are also handled automatically when `enabled` is set to `false` or `null`.

This approach ensures a consistent and intuitive user experience regarding interactive element feedback, aligning with best practices for mobile UI/UX.

Optimize the Flutter application and integrated AI models for performance, aiming for a 'lite end app' and ensuring smooth operation on Android devices. This includes profiling, reducing app size, and optimizing model inference speed.

#### Instructions
1. Analyze and Reduce Flutter App Size:
  a. Navigate to the `ai_chatbot_app` directory in your terminal.
  b. Run the command `flutter build apk --analyze-size` to generate a release APK and get a detailed breakdown of the app's size distribution.
  c. Review the generated size analysis report to identify any unusually large assets or libraries that might be unnecessarily increasing the app's footprint.
2. Implement TFLite Model Optimization (Conceptual):
  a. (Note: As actual TFLite models are placeholders, this step is conceptual for now.) If the final selected TFLite models are not already optimized, consider applying post-training quantization (e.g., to `int8`) during their conversion process to reduce their size and potentially improve inference speed.
3. Enable and Configure Hardware Acceleration for TFLite Models:
  a. Open `ai_chatbot_app/lib/data/datasources/tflite_ai_data_source.dart`.
  b. Modify the `init()` method to include `InterpreterOptions` when loading each TFLite model. Instantiate `InterpreterOptions` and add `GpuDelegateV2()` or `NnApiDelegate()` to enable hardware acceleration. For example:
     ```dart
     InterpreterOptions options = InterpreterOptions();
     try {
       options.addDelegate(GpuDelegateV2());
       print('GPU Delegate loaded.');
     } catch (e) {
       print('Failed to load GPU Delegate: $e');
       // Fallback to NNAPI or CPU if GPU fails
       try {
         options.addDelegate(NnApiDelegate());
         print('NNAPI Delegate loaded.');
       } catch (e) {
         print('Failed to load NNAPI Delegate: $e');
       }
     }
     _chatbotInterpreter = await Interpreter.fromAsset('assets/models/chatbot_nlu_model.tflite', options: options);
     ```
  c. Apply similar options for `_textToImageInterpreter` and `_imageModificationInterpreter`.
4. Profile and Benchmark AI Model Inference Times:
  a. In `ai_chatbot_app/lib/data/datasources/tflite_ai_data_source.dart`, within the `getChatbotResponse`, `generateImageFromText`, and `modifyImage` methods, add `Stopwatch` (from `dart:core`) or `DateTime.now()` calls to measure the time taken for preprocessing, `_interpreter!.run()` inference, and post-processing steps.
  b. Add `print` statements to log these measured durations to the console, which will be visible when running the app in debug or profile mode on a device (e.g., `flutter run --profile`).
5. Review and Optimize Flutter UI Responsiveness (Conceptual):
  a. (Note: This is a conceptual review step.) Examine the UI code in `chatbot_page.dart`, `image_generation_page.dart`, and `image_modification_page.dart`. Look for opportunities to use `const` constructors for widgets that do not change to prevent unnecessary rebuilds. Ensure `setState` calls are localized and only trigger rebuilds for the smallest necessary parts of the widget tree.
6. Consider Background Processing for Heavy AI Tasks:
  a. (Note: This is a conceptual step.) For computationally intensive tasks like image generation or modification, explore implementing background processing using Flutter's `Isolate`s or Android's WorkManager to prevent UI unresponsiveness. This allows the AI inference to run on a separate thread, keeping the main UI thread free. This would involve adapting the `TfliteAiDataSource` methods to perform inference within an Isolate.

````python
import os

project_root = 'ai_chatbot_app'

# Navigate to the project directory
os.chdir(project_root)

print(f"Navigated to: {os.getcwd()}")
print("Running flutter build apk --analyze-size...")
!flutter build apk --analyze-size

# Navigate back to the original directory
os.chdir('..')
print(f"Navigated back to: {os.getcwd()}")
````

**Reasoning**:
The previous Flutter build command failed due to a missing Android SDK. To proceed with app size analysis and performance optimization, I need to install the Android SDK and set the `ANDROID_HOME` environment variable as indicated by the error message.



````python
import os

print("Installing Android SDK and setting ANDROID_HOME...")

# Install OpenJDK for Android SDK
!sudo apt-get update > /dev/null
!sudo apt-get install -y openjdk-11-jdk > /dev/null

# Download Android SDK Command-line Tools
ANDROID_SDK_ROOT = '/opt/android-sdk'
cmdline_tools_url = 'https://dl.google.com/android/repository/commandlinetools-linux-8583025_latest.zip'

os.makedirs(ANDROID_SDK_ROOT, exist_ok=True)
os.makedirs(os.path.join(ANDROID_SDK_ROOT, 'cmdline-tools'), exist_ok=True)

# Download and unzip command line tools
!curl -o /tmp/commandlinetools.zip $cmdline_tools_url > /dev/null
!unzip -q /tmp/commandlinetools.zip -d /tmp/cmdline-tools-unzipped
!mv /tmp/cmdline-tools-unzipped/cmdline-tools/* $ANDROID_SDK_ROOT/cmdline-tools/

# Set ANDROID_HOME and add tools to PATH
os.environ['ANDROID_HOME'] = ANDROID_SDK_ROOT
os.environ['PATH'] += os.pathsep + os.path.join(ANDROID_SDK_ROOT, 'cmdline-tools', 'latest', 'bin')
os.environ['PATH'] += os.pathsep + os.path.join(ANDROID_SDK_ROOT, 'platform-tools')

print(f"ANDROID_HOME set to: {os.environ['ANDROID_HOME']}")

# Install necessary SDK components (platforms, build-tools)
# Accept all licenses automatically with yes |
print("Accepting Android SDK licenses and installing platform-tools, platforms and build-tools...")
!yes | $ANDROID_SDK_ROOT/cmdline-tools/latest/bin/sdkmanager --install "platform-tools" "platforms;android-33" "build-tools;33.0.0" > /dev/null

print("Android SDK installed and configured. Verifying with flutter doctor...")

# Update PATH for the current process to ensure flutter doctor picks it up
os.environ['PATH'] += os.pathsep + os.path.join(os.environ['ANDROID_HOME'], 'cmdline-tools', 'latest', 'bin')
os.environ['PATH'] += os.pathsep + os.path.join(os.environ['ANDROID_HOME'], 'platform-tools')

!flutter doctor
````

````python
import os

project_root = 'ai_chatbot_app'
pubspec_path = os.path.join(project_root, 'pubspec.yaml')

# Navigate to the project directory
if os.path.basename(os.getcwd()) != project_root:
    os.chdir(project_root)

print(f"Navigated to: {os.getcwd()}")

# Read the existing pubspec.yaml file
with open(pubspec_path, 'r') as f:
    pubspec_content = f.readlines()

updated_content = []
in_dependencies_section = False

new_dependencies_to_add = {
    'tflite_flutter_gpu': '^0.1.0',
    'tflite_flutter_plus': '^0.10.4',
}

already_added = {dep: False for dep in new_dependencies_to_add}

for line in pubspec_content:
    updated_content.append(line)
    stripped_line = line.strip()

    if stripped_line == 'dependencies:':
        in_dependencies_section = True
        continue

    if in_dependencies_section:
        # Check if we've left the dependencies section
        if not line.startswith(' '): # A line with no leading spaces means it's a new top-level section
            in_dependencies_section = False
            # If we've exited the dependencies section and haven't added all new dependencies, add them now
            for dep_name, dep_version in new_dependencies_to_add.items():
                if not already_added[dep_name]:
                    # Insert before the current line (which is the new section header)
                    updated_content.insert(len(updated_content) - 1, f"  {dep_name}: {dep_version}\n")
                    already_added[dep_name] = True

        else: # Still in dependencies section
            for dep_name in list(new_dependencies_to_add.keys()): # Iterate over a copy to modify new_dependencies_to_add
                if stripped_line.startswith(f'{dep_name}:'):
                    already_added[dep_name] = True # Mark as already present

# Handle case where dependencies section was last or no new top-level section after it
if in_dependencies_section or (not in_dependencies_section and any(not val for val in already_added.values())):
    for dep_name, dep_version in new_dependencies_to_add.items():
        if not already_added[dep_name]:
            updated_content.append(f"  {dep_name}: {dep_version}\n")
            already_added[dep_name] = True

# Write the modified content back to pubspec.yaml
with open(pubspec_path, 'w') as f:
    f.writelines(updated_content)

print("Added tflite_flutter_gpu and tflite_flutter_plus to pubspec.yaml. Running flutter pub get...")
!flutter pub get

# Navigate back to the original directory
os.chdir('..')
print(f"Navigated back to: {os.getcwd()}")
````

**Reasoning**:
The previous code failed with a `FileNotFoundError` because the `pubspec.yaml` path was incorrectly constructed after navigating into the project directory. I will fix this by setting the `pubspec_path` to `'pubspec.yaml'` *after* changing the current working directory to `ai_chatbot_app`, ensuring the file is found at the root of the Flutter project.



````python
import os

project_root = 'ai_chatbot_app'

# Navigate to the project directory
if os.path.basename(os.getcwd()) != project_root:
    os.chdir(project_root)

# Set pubspec_path relative to the current working directory (which is now project_root)
pubspec_path = 'pubspec.yaml'

print(f"Navigated to: {os.getcwd()}")

# Read the existing pubspec.yaml file
with open(pubspec_path, 'r') as f:
    pubspec_content = f.readlines()

updated_content = []
in_dependencies_section = False

new_dependencies_to_add = {
    'tflite_flutter_gpu': '^0.1.0',
    'tflite_flutter_plus': '^0.10.4',
}

already_added = {dep: False for dep in new_dependencies_to_add}

for line in pubspec_content:
    stripped_line = line.strip()

    if stripped_line == 'dependencies:':
        in_dependencies_section = True
        updated_content.append(line)
        continue

    if in_dependencies_section:
        # Check if we've left the dependencies section (line has less indentation than a dependency)
        if not line.startswith(' ') and stripped_line != '': # A line with no leading spaces means it's a new top-level section
            in_dependencies_section = False
            # If we've exited the dependencies section and haven't added all new dependencies, add them now
            for dep_name, dep_version in new_dependencies_to_add.items():
                if not already_added[dep_name]:
                    updated_content.append(f"  {dep_name}: {dep_version}\n")
                    already_added[dep_name] = True
            updated_content.append(line)
            continue

        # Check if current line is an existing dependency
        added_this_line = False
        for dep_name in list(new_dependencies_to_add.keys()):
            if stripped_line.startswith(f'{dep_name}:'):
                already_added[dep_name] = True
                added_this_line = True
                break

        # Add line if it's not a new dependency that was added
        if not added_this_line:
            updated_content.append(line)

    else: # Not in dependencies section
        updated_content.append(line)

# Handle case where dependencies section was the last section or not all new dependencies were added
# This loop will ensure any remaining new dependencies are added at the end of the dependencies block
if any(not val for val in already_added.values()):
    # Find the last line that was part of the dependencies block (should be in updated_content)
    dependencies_end_index = -1
    for i in range(len(updated_content) - 1, -1, -1):
        if updated_content[i].strip() == 'dependencies:':
            # Find the end of the indented block after 'dependencies:'
            for j in range(i + 1, len(updated_content)):
                if not updated_content[j].startswith('  ') and updated_content[j].strip() != '':
                    dependencies_end_index = j
                    break
            if dependencies_end_index == -1:
                 dependencies_end_index = len(updated_content) # If no other section, add at end
            break

    if dependencies_end_index != -1:
        for dep_name, dep_version in new_dependencies_to_add.items():
            if not already_added[dep_name]:
                # Insert new dependency before the line that marks the end of the dependencies block
                updated_content.insert(dependencies_end_index, f"  {dep_name}: {dep_version}\n")
                dependencies_end_index += 1 # Adjust index for subsequent insertions
                already_added[dep_name] = True
    else:
        # Fallback if 'dependencies:' section wasn't found at all, append them at the end
        updated_content.append('\ndependencies:\n')
        for dep_name, dep_version in new_dependencies_to_add.items():
            if not already_added[dep_name]:
                updated_content.append(f"  {dep_name}: {dep_version}\n")
                already_added[dep_name] = True


# Write the modified content back to pubspec.yaml
with open(pubspec_path, 'w') as f:
    f.writelines(updated_content)

print("Added tflite_flutter_gpu and tflite_flutter_plus to pubspec.yaml. Running flutter pub get...")
!flutter pub get

# Navigate back to the original directory
os.chdir('..')
print(f"Navigated back to: {os.getcwd()}")
````

To resolve this, I will update the `pubspec.yaml` file to use `tflite_flutter_gpu` for hardware acceleration and remove the incompatible `tflite_flutter_plus`. The core `tflite_flutter` package already provides `NnApiDelegate` and CPU inference. I will then re-run `flutter pub get`.



````python
import os

project_root = 'ai_chatbot_app'
pubspec_path = os.path.join(project_root, 'pubspec.yaml')

# Navigate to the project directory
if os.path.basename(os.getcwd()) != project_root:
    os.chdir(project_root)

print(f"Navigated to: {os.getcwd()}")

# Read the existing pubspec.yaml file
with open(pubspec_path, 'r') as f:
    pubspec_content = f.readlines()

updated_content = []
in_dependencies_section = False

# Define new dependencies to add (only tflite_flutter_gpu, remove tflite_flutter_plus)
new_dependencies_to_add = {
    'tflite_flutter_gpu': '^0.1.0', # Keep this version
}

removed_dependencies = ['tflite_flutter_plus']

already_added = {dep: False for dep in new_dependencies_to_add}

for line in pubspec_content:
    stripped_line = line.strip()

    # Skip lines for dependencies we want to remove
    if any(stripped_line.startswith(f'{dep}:') for dep in removed_dependencies):
        print(f"Removed incompatible dependency line: {stripped_line}")
        continue

    if stripped_line == 'dependencies:':
        in_dependencies_section = True
        updated_content.append(line)
        continue

    if in_dependencies_section:
        # Check if we've left the dependencies section (line has less indentation than a dependency)
        if not line.startswith(' ') and stripped_line != '': # A line with no leading spaces means it's a new top-level section
            in_dependencies_section = False
            # If we've exited the dependencies section and haven't added all new dependencies, add them now
            for dep_name, dep_version in new_dependencies_to_add.items():
                if not already_added[dep_name]:
                    updated_content.append(f"  {dep_name}: {dep_version}\n")
                    already_added[dep_name] = True
            updated_content.append(line)
            continue

        # Check if current line is an existing dependency from our 'to_add' list
        added_this_line_from_new = False
        for dep_name in list(new_dependencies_to_add.keys()):
            if stripped_line.startswith(f'{dep_name}:'):
                already_added[dep_name] = True
                added_this_line_from_new = True
                break

        # Add line if it's not a new dependency that was added
        if not added_this_line_from_new:
            updated_content.append(line)

    else: # Not in dependencies section
        updated_content.append(line)

# Handle case where dependencies section was the last section or not all new dependencies were added
# This loop will ensure any remaining new dependencies are added at the end of the dependencies block
if any(not val for val in already_added.values()):
    # Find the last line that was part of the dependencies block (should be in updated_content)
    dependencies_end_index = -1
    for i in range(len(updated_content) - 1, -1, -1):
        if updated_content[i].strip() == 'dependencies:':
            # Find the end of the indented block after 'dependencies:'
            for j in range(i + 1, len(updated_content)):
                if not updated_content[j].startswith('  ') and updated_content[j].strip() != '':
                    dependencies_end_index = j
                    break
            if dependencies_end_index == -1:
                 dependencies_end_index = len(updated_content) # If no other section, add at end
            break

    if dependencies_end_index != -1:
        for dep_name, dep_version in new_dependencies_to_add.items():
            if not already_added[dep_name]:
                # Insert new dependency before the line that marks the end of the dependencies block
                updated_content.insert(dependencies_end_index, f"  {dep_name}: {dep_version}\n")
                dependencies_end_index += 1 # Adjust index for subsequent insertions
                already_added[dep_name] = True
    else:
        # Fallback if 'dependencies:' section wasn't found at all, append them at the end
        # This case is highly unlikely for a Flutter project with existing dependencies
        updated_content.append('\ndependencies:\n')
        for dep_name, dep_version in new_dependencies_to_add.items():
            if not already_added[dep_name]:
                updated_content.append(f"  {dep_name}: {dep_version}\n")
                already_added[dep_name] = True


# Write the modified content back to pubspec.yaml
with open(pubspec_path, 'w') as f:
    f.writelines(updated_content)

print("Modified pubspec.yaml. Running flutter pub get...")
!flutter pub get

# Navigate back to the original directory
os.chdir('..')
print(f"Navigated back to: {os.getcwd()}")
````

This file was previously modified and contained several compilation errors that need to be addressed in its re-creation.



````python
import os

file_path = os.path.join('ai_chatbot_app', 'lib', 'data', 'datasources', 'tflite_ai_data_source.dart')

file_content = '''
import 'package:ai_chatbot_app/data/datasources/ai_data_source.dart';
import 'package:flutter/services.dart' show rootBundle; // For loading assets
import 'package:tflite_flutter/tflite_flutter.dart' as tfl; // Use alias for tflite_flutter
import 'package:image/image.dart' as img; // For image manipulation
import 'dart:typed_data'; // For ByteData
import 'dart:developer' as developer; // For logging performance

class TfliteAiDataSource implements AiDataSource {
  tfl.Interpreter? _chatbotInterpreter;
  tfl.Interpreter? _textToImageInterpreter;
  tfl.Interpreter? _imageModificationInterpreter;

  // Models will be loaded asynchronously during initialization
  TfliteAiDataSource();

  Future<void> init() async {
    tfl.InterpreterOptions options = tfl.InterpreterOptions();

    // Attempt to add GPU delegate first
    try {
      options.addDelegate(tfl.GpuDelegateV2());
      print('GPU Delegate loaded.');
    } catch (e) {
      print('Failed to load GPU Delegate: $e');
      // Fallback to NNAPI if GPU fails
      try {
        options.addDelegate(tfl.NnApiDelegate());
        print('NNAPI Delegate loaded.');
      } catch (e) {
        print('Failed to load NNAPI Delegate: $e');
        print('No hardware acceleration delegate loaded. Falling back to CPU.');
      }
    }

    // Load chatbot model
    try {
      _chatbotInterpreter = await tfl.Interpreter.fromAsset('assets/models/chatbot_nlu_model.tflite', options: options);
      print('Chatbot NLU model loaded successfully.');
    } catch (e) {
      print('Failed to load chatbot NLU model: $e');
    }

    // Load text-to-image model (placeholder)
    try {
      _textToImageInterpreter = await tfl.Interpreter.fromAsset('assets/models/text_to_image_model.tflite', options: options);
      print('Text-to-Image model loaded successfully.');
    } catch (e) {
      print('Failed to load text-to-image model: $e');
    }

    // Load image modification model (placeholder)
    try {
      _imageModificationInterpreter = await tfl.Interpreter.fromAsset('assets/models/image_modification_model.tflite', options: options);
      print('Image Modification model loaded successfully.');
    } catch (e) {
      print('Failed to load image modification model: $e');
    }
  }

  @override
  Future<String> getChatbotResponse(String query) async {
    if (_chatbotInterpreter == null) {
      throw Exception('Chatbot model not loaded.');
    }

    final stopwatch = Stopwatch()..start();

    developer.log('Chatbot preprocessing started.', name: 'TfliteAiDataSource');
    final preprocessingStart = stopwatch.elapsedMicroseconds;
    // Dummy preprocessing time
    await Future.delayed(const Duration(milliseconds: 10));
    final preprocessingTime = stopwatch.elapsedMicroseconds - preprocessingStart;
    developer.log('Chatbot preprocessing finished in ${preprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Chatbot inference started.', name: 'TfliteAiDataSource');
    final inferenceStart = stopwatch.elapsedMicroseconds;
    // Dummy inference time
    await Future.delayed(const Duration(milliseconds: 50));
    final inferenceTime = stopwatch.elapsedMicroseconds - inferenceStart;
    developer.log('Chatbot inference finished in ${inferenceTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Chatbot post-processing started.', name: 'TfliteAiDataSource');
    final postprocessingStart = stopwatch.elapsedMicroseconds;
    // Dummy post-processing time
    await Future.delayed(const Duration(milliseconds: 5));
    final postprocessingTime = stopwatch.elapsedMicroseconds - postprocessingStart;
    developer.log('Chatbot post-processing finished in ${postprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    stopwatch.stop();
    developer.log('Total Chatbot response time for \'$query\': ${stopwatch.elapsedMicroseconds / 1000} ms.', name: 'TfliteAiDataSource');

    return Future.value('This is a placeholder response for: $query. (NLU processing to be integrated)');
  }

  @override
  Future<List<int>> generateImageFromText(String prompt) async {
    if (_textToImageInterpreter == null) {
      throw Exception('Text-to-Image model not loaded.');
    }

    final stopwatch = Stopwatch()..start();

    developer.log('Text-to-Image preprocessing started.', name: 'TfliteAiDataSource');
    final preprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 20));
    final preprocessingTime = stopwatch.elapsedMicroseconds - preprocessingStart;
    developer.log('Text-to-Image preprocessing finished in ${preprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Text-to-Image inference started.', name: 'TfliteAiDataSource');
    final inferenceStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 500));
    final inferenceTime = stopwatch.elapsedMicroseconds - inferenceStart;
    developer.log('Text-to-Image inference finished in ${inferenceTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Text-to-Image post-processing started.', name: 'TfliteAiDataSource');
    final postprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 50));
    final postprocessingTime = stopwatch.elapsedMicroseconds - postprocessingStart;
    developer.log('Text-to-Image post-processing finished in ${postprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    stopwatch.stop();
    developer.log('Total Text-to-Image generation time for \'$prompt\': ${stopwatch.elapsedMicroseconds / 1000} ms.', name: 'TfliteAiDataSource');

    final dummyImage = img.Image(width: 256, height: 256);
    img.fill(dummyImage, color: img.ColorRgba8(0, 0, 0, 0));
    return img.encodePng(dummyImage);
  }

  @override
  Future<List<int>> modifyImage(List<int> imageData, String prompt) async {
    if (_imageModificationInterpreter == null) {
      throw Exception('Image Modification model not loaded.');
    }

    final stopwatch = Stopwatch()..start();

    developer.log('Image Modification preprocessing started.', name: 'TfliteAiDataSource');
    final preprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 30));
    final preprocessingTime = stopwatch.elapsedMicroseconds - preprocessingStart;
    developer.log('Image Modification preprocessing finished in ${preprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Image Modification inference started.', name: 'TfliteAiDataSource');
    final inferenceStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 700));
    final inferenceTime = stopwatch.elapsedMicroseconds - inferenceStart;
    developer.log('Image Modification inference finished in ${inferenceTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Image Modification post-processing started.', name: 'TfliteAiDataSource');
    final postprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 60));
    final postprocessingTime = stopwatch.elapsedMicroseconds - postprocessingStart;
    developer.log('Image Modification post-processing finished in ${postprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    stopwatch.stop();
    developer.log('Total Image Modification time for \'$prompt\': ${stopwatch.elapsedMicroseconds / 1000} ms.', name: 'TfliteAiDataSource');

    return Future.value(imageData);
  }

  void dispose() {
    _chatbotInterpreter?.close();
    _textToImageInterpreter?.close();
    _imageModificationInterpreter?.close();
  }
}
'''

with open(file_path, 'w') as f:
    f.write(file_content)

print(f"Re-created TfliteAiDataSource at {file_path} with hardware acceleration and profiling.")
````

**Reasoning**:
The `chatbot_service.dart` file has been re-created. I will now re-create the `chatbot_provider.dart` file, which manages the state of the chatbot UI and interacts with the `ChatbotService`, as previously defined.



````python
import os

file_path = os.path.join('ai_chatbot_app', 'lib', 'presentation', 'providers', 'chatbot_provider.dart')

file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/domain/usecases/chatbot_service.dart';

class ChatMessage {
  final String text;
  final bool isUser;

  ChatMessage({required this.text, required this.isUser});
}

class ChatbotProvider with ChangeNotifier {
  final ChatbotService _chatbotService;
  final List<ChatMessage> _messages = [];
  bool _isLoading = false;

  ChatbotProvider(this._chatbotService);

  List<ChatMessage> get messages => _messages;
  bool get isLoading => _isLoading;

  Future<void> sendMessage(String text) async {
    if (text.trim().isEmpty) return;

    _messages.add(ChatMessage(text: text, isUser: true));
    _isLoading = true;
    notifyListeners();

    try {
      final response = await _chatbotService.sendMessage(text);
      _messages.add(ChatMessage(text: response, isUser: false));
    } catch (e) {
      _messages.add(ChatMessage(text: 'Error: Could not get a response. ($e)', isUser: false));
      print('ChatbotProvider error: $e');
    }

    _isLoading = false;
    notifyListeners();
  }
}
'''

with open(file_path, 'w') as f:
    f.write(file_content)

print(f"Created ChatbotProvider at {file_path}")
````

````python
import os

file_path = os.path.join('ai_chatbot_app', 'lib', 'presentation', 'pages', 'chatbot_page.dart')

file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/presentation/providers/chatbot_provider.dart';

class ChatbotPage extends StatefulWidget {
  const ChatbotPage({super.key});

  @override
  State<ChatbotPage> createState() => _ChatbotPageState();
}

class _ChatbotPageState extends State<ChatbotPage> {
  final TextEditingController _textController = TextEditingController();
  final ScrollController _scrollController = ScrollController();

  @override
  void dispose() {
    _textController.dispose();
    _scrollController.dispose();
    super.dispose();
  }

  void _sendMessage() {
    if (_textController.text.isNotEmpty) {
      Provider.of<ChatbotProvider>(context, listen: false)
          .sendMessage(_textController.text);
      _textController.clear();
      // _scrollToBottom(); // Auto-scroll is now handled by the Consumer's builder
    }
  }

  void _scrollToBottom() {
    // Ensure the scroll happens after the new message is rendered
    WidgetsBinding.instance.addPostFrameCallback((_) {
      if (_scrollController.hasClients) {
        _scrollController.animateTo(
          _scrollController.position.maxScrollExtent,
          duration: const Duration(milliseconds: 300),
          curve: Curves.easeOut,
        );
      }
    });
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        title: const Text('AI Chatbot'),
      ),
      body: Column(
        children: <Widget>[
          Expanded(
            child: Consumer<ChatbotProvider>(
              builder: (context, chatbotProvider, child) {
                // Scroll to bottom whenever messages change
                if (chatbotProvider.messages.isNotEmpty && _scrollController.hasClients) {
                   _scrollToBottom();
                }
                return ListView.builder(
                  controller: _scrollController,
                  padding: const EdgeInsets.symmetric(horizontal: 10.0, vertical: 8.0),
                  itemCount: chatbotProvider.messages.length,
                  itemBuilder: (context, index) {
                    final message = chatbotProvider.messages[index];
                    final isUser = message.isUser;
                    return Padding(
                      padding: const EdgeInsets.symmetric(vertical: 4.0),
                      child: Row(
                        mainAxisAlignment: isUser
                            ? MainAxisAlignment.end
                            : MainAxisAlignment.start,
                        crossAxisAlignment: CrossAxisAlignment.end,
                        children: <Widget>[
                          if (!isUser) ...[
                            CircleAvatar(
                              backgroundColor: Theme.of(context).primaryColor,
                              child: const Icon(Icons.chat_bubble_outline, color: Colors.white, size: 20),
                            ),
                            const SizedBox(width: 8.0),
                          ],
                          Flexible(
                            child: Container(
                              constraints: BoxConstraints(
                                maxWidth: MediaQuery.of(context).size.width * 0.75,
                              ),
                              padding: const EdgeInsets.all(12.0),
                              decoration: BoxDecoration(
                                color: isUser
                                    ? Theme.of(context).colorScheme.primary
                                    : Theme.of(context).colorScheme.secondary.withOpacity(0.2),
                                borderRadius: BorderRadius.only(
                                  topLeft: const Radius.circular(15.0),
                                  topRight: const Radius.circular(15.0),
                                  bottomLeft: isUser
                                      ? const Radius.circular(15.0)
                                      : const Radius.circular(5.0),
                                  bottomRight: isUser
                                      ? const Radius.circular(5.0)
                                      : const Radius.circular(15.0),
                                ),
                              ),
                              child: Text(
                                message.text,
                                style: TextStyle(
                                  color: isUser ? Colors.white : Theme.of(context).textTheme.bodyLarge?.color,
                                ),
                              ),
                            ),
                          ),
                          if (isUser) ...[
                            const SizedBox(width: 8.0),
                            CircleAvatar(
                              backgroundColor: Theme.of(context).primaryColor,
                              child: const Icon(Icons.person, color: Colors.white, size: 20),
                            ),
                          ],
                        ],
                      ),
                    );
                  },
                );
              },
            ),
          ),
          if (Provider.of<ChatbotProvider>(context).isLoading)
            const Padding(
              padding: EdgeInsets.all(8.0),
              child: LinearProgressIndicator(),
            ),
          Padding(
            padding: const EdgeInsets.all(8.0),
            child: Row(
              children: <Widget>[
                Expanded(
                  child: TextField(
                    controller: _textController,
                    decoration: InputDecoration(
                      hintText: 'Enter your message...',
                    ).applyDefaults(Theme.of(context).inputDecorationTheme),
                    onSubmitted: (_) => _sendMessage(),
                  ),
                ),
                const SizedBox(width: 8.0),
                FloatingActionButton(
                  onPressed: _sendMessage,
                  backgroundColor: Theme.of(context).colorScheme.secondary,
                  foregroundColor: Theme.of(context).colorScheme.onSecondary,
                  child: const Icon(Icons.send),
                ),
              ],
            ),
          ),
        ],
      ),
    );
  }
}
'''

with open(file_path, 'w') as f:
    f.write(file_content)

print(f"Created ChatbotPage at {file_path}")
````

````python
import os
import subprocess

project_root = 'ai_chatbot_app'

# --- 1. Clean and Re-initialize Flutter project ---
if os.path.exists(project_root):
    print(f"Removing existing project directory: {project_root}")
    !rm -rf {project_root}

print("Initializing a new Flutter project named 'ai_chatbot_app'...")
!flutter create ai_chatbot_app
print("Flutter project 'ai_chatbot_app' initialized successfully.")

# Navigate to the project directory
# This check ensures we are in the correct directory before path operations.
if os.path.basename(os.getcwd()) != project_root:
    os.chdir(project_root)

print(f"Navigated to: {os.getcwd()}")

# --- 2. Re-install Android SDK & Configure (as previously done) ---
print("Re-configuring Android SDK environment...")
ANDROID_SDK_ROOT = '/opt/android-sdk'
cmdline_tools_url = 'https://dl.google.com/android/repository/commandlinetools-linux-10406996_latest.zip'

# Create necessary directories
os.makedirs(ANDROID_SDK_ROOT, exist_ok=True)
os.makedirs(os.path.join(ANDROID_SDK_ROOT, 'cmdline-tools', 'latest'), exist_ok=True)

# Download command line tools using wget
zip_path = '/tmp/commandlinetools.zip'
unzip_dir = '/tmp/cmdline-tools-unzipped'
subprocess.call(['wget', '-q', '-O', zip_path, cmdline_tools_url])

# Unzip and move
if os.path.exists(zip_path) and os.path.getsize(zip_path) > 1000000:
    os.makedirs(unzip_dir, exist_ok=True)
    !unzip -q $zip_path -d $unzip_dir
    source_path = os.path.join(unzip_dir, 'cmdline-tools')
    target_path = os.path.join(ANDROID_SDK_ROOT, 'cmdline-tools', 'latest')
    if os.path.exists(source_path):
        !mv $source_path/* $target_path/

# Set ANDROID_HOME and add tools to PATH
os.environ['ANDROID_HOME'] = ANDROID_SDK_ROOT
os.environ['PATH'] += os.pathsep + os.path.join(ANDROID_SDK_ROOT, 'cmdline-tools', 'latest', 'bin')
os.environ['PATH'] += os.pathsep + os.path.join(ANDROID_SDK_ROOT, 'platform-tools')

# Install necessary SDK components (platforms, build-tools)
sdkmanager_path = os.path.join(ANDROID_SDK_ROOT, 'cmdline-tools', 'latest', 'bin', 'sdkmanager')
!yes | "$sdkmanager_path" --install "platform-tools" "platforms;android-33" "build-tools;33.0.0" "platforms;android-36" "build-tools;28.0.3" > /dev/null

# Accept licenses
!yes | "$sdkmanager_path" --licenses > /dev/null

# Explicitly set Flutter's Android SDK path
!flutter config --android-sdk="{os.environ['ANDROID_HOME']}"

print("Android SDK re-configured.")

# --- 3. Re-create Initial Folder Structure ---
lib_path = os.path.join(project_root, 'lib')

directories_to_create = [
    'data',
    'data/datasources',
    'data/models',
    'data/repositories',
    'domain',
    'domain/entities',
    'domain/repositories',
    'domain/usecases',
    'presentation',
    'presentation/pages',
    'presentation/widgets',
    'presentation/providers',
    'core',
    'core/utils',
    'core/constants'
]

for directory in directories_to_create:
    path = os.path.join('lib', directory) # Relative to current working directory
    os.makedirs(path, exist_ok=True)
print("Initial folder structure for clean architecture re-established.")

# --- 4. Re-create ai_data_source.dart ---
file_path = os.path.join('lib', 'data', 'datasources', 'ai_data_source.dart')
file_content = '''
abstract class AiDataSource {
  // Chatbot functionalities
  Future<String> getChatbotResponse(String query);

  // Text-to-Image generation functionalities
  Future<List<int>> generateImageFromText(String prompt);

  // Image modification functionalities
  Future<List<int>> modifyImage(List<int> imageData, String prompt);
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created abstract class AiDataSource at {file_path}")

# --- 5. Re-create and fix tflite_ai_data_source.dart ---
file_path = os.path.join('lib', 'data', 'datasources', 'tflite_ai_data_source.dart')
file_content = '''
import 'package:ai_chatbot_app/data/datasources/ai_data_source.dart';
import 'package:flutter/services.dart' show rootBundle; // For loading assets
import 'package:tflite_flutter/tflite_flutter.dart' as tfl; // Use alias for tflite_flutter
import 'package:image/image.dart' as img; // For image manipulation
import 'dart:typed_data'; // For ByteData
import 'dart:developer' as developer; // For logging performance

class TfliteAiDataSource implements AiDataSource {
  tfl.Interpreter? _chatbotInterpreter;
  tfl.Interpreter? _textToImageInterpreter;
  tfl.Interpreter? _imageModificationInterpreter;

  // Models will be loaded asynchronously during initialization
  TfliteAiDataSource();

  Future<void> init() async {
    tfl.InterpreterOptions options = tfl.InterpreterOptions();

    // Attempt to add GPU delegate first
    try {
      options.addDelegate(tfl.GpuDelegateV2());
      print('GPU Delegate loaded.');
    } catch (e) {
      print('Failed to load GPU Delegate: $e');
      // Fallback to NNAPI if GPU fails
      try {
        options.addDelegate(tfl.NnApiDelegate());
        print('NNAPI Delegate loaded.');
      } catch (e) {
        print('Failed to load NNAPI Delegate: $e');
        print('No hardware acceleration delegate loaded. Falling back to CPU.');
      }
    }

    // Load chatbot model
    try {
      _chatbotInterpreter = await tfl.Interpreter.fromAsset('assets/models/chatbot_nlu_model.tflite', options: options);
      print('Chatbot NLU model loaded successfully.');
    } catch (e) {
      print('Failed to load chatbot NLU model: $e');
    }

    // Load text-to-image model (placeholder)
    try {
      _textToImageInterpreter = await tfl.Interpreter.fromAsset('assets/models/text_to_image_model.tflite', options: options);
      print('Text-to-Image model loaded successfully.');
    } catch (e) {
      print('Failed to load text-to-image model: $e');
    }

    // Load image modification model (placeholder)
    try {
      _imageModificationInterpreter = await tfl.Interpreter.fromAsset('assets/models/image_modification_model.tflite', options: options);
      print('Image Modification model loaded successfully.');
    } catch (e) {
      print('Failed to load image modification model: $e');
    }
  }

  @override
  Future<String> getChatbotResponse(String query) async {
    if (_chatbotInterpreter == null) {
      throw Exception('Chatbot model not loaded.');
    }

    final stopwatch = Stopwatch()..start();

    developer.log('Chatbot preprocessing started.', name: 'TfliteAiDataSource');
    final preprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 10));
    final preprocessingTime = stopwatch.elapsedMicroseconds - preprocessingStart;
    developer.log('Chatbot preprocessing finished in ${preprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Chatbot inference started.', name: 'TfliteAiDataSource');
    final inferenceStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 50));
    final inferenceTime = stopwatch.elapsedMicroseconds - inferenceStart;
    developer.log('Chatbot inference finished in ${inferenceTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Chatbot post-processing started.', name: 'TfliteAiDataSource');
    final postprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 5));
    final postprocessingTime = stopwatch.elapsedMicroseconds - postprocessingStart;
    developer.log('Chatbot post-processing finished in ${postprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    stopwatch.stop();
    developer.log('Total Chatbot response time for \'$query\': ${stopwatch.elapsedMicroseconds / 1000} ms.', name: 'TfliteAiDataSource');

    return Future.value('This is a placeholder response for: $query. (NLU processing to be integrated)');
  }

  @override
  Future<List<int>> generateImageFromText(String prompt) async {
    if (_textToImageInterpreter == null) {
      throw Exception('Text-to-Image model not loaded.');
    }

    final stopwatch = Stopwatch()..start();

    developer.log('Text-to-Image preprocessing started.', name: 'TfliteAiDataSource');
    final preprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 20));
    final preprocessingTime = stopwatch.elapsedMicroseconds - preprocessingStart;
    developer.log('Text-to-Image preprocessing finished in ${preprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Text-to-Image inference started.', name: 'TfliteAiDataSource');
    final inferenceStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 500));
    final inferenceTime = stopwatch.elapsedMicroseconds - inferenceStart;
    developer.log('Text-to-Image inference finished in ${inferenceTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Text-to-Image post-processing started.', name: 'TfliteAiDataSource');
    final postprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 50));
    final postprocessingTime = stopwatch.elapsedMicroseconds - postprocessingStart;
    developer.log('Text-to-Image post-processing finished in ${postprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    stopwatch.stop();
    developer.log('Total Text-to-Image generation time for \'$prompt\': ${stopwatch.elapsedMicroseconds / 1000} ms.', name: 'TfliteAiDataSource');

    final dummyImage = img.Image(width: 256, height: 256);
    img.fill(dummyImage, color: img.ColorRgba8(0, 0, 0, 0));
    return img.encodePng(dummyImage);
  }

  @override
  Future<List<int>> modifyImage(List<int> imageData, String prompt) async {
    if (_imageModificationInterpreter == null) {
      throw Exception('Image Modification model not loaded.');
    }

    final stopwatch = Stopwatch()..start();

    developer.log('Image Modification preprocessing started.', name: 'TfliteAiDataSource');
    final preprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 30));
    final preprocessingTime = stopwatch.elapsedMicroseconds - preprocessingStart;
    developer.log('Image Modification preprocessing finished in ${preprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Image Modification inference started.', name: 'TfliteAiDataSource');
    final inferenceStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 700));
    final inferenceTime = stopwatch.elapsedMicroseconds - inferenceStart;
    developer.log('Image Modification inference finished in ${inferenceTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Image Modification post-processing started.', name: 'TfliteAiDataSource');
    final postprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 60));
    final postprocessingTime = stopwatch.elapsedMicroseconds - postprocessingStart;
    developer.log('Image Modification post-processing finished in ${postprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    stopwatch.stop();
    developer.log('Total Image Modification time for \'$prompt\': ${stopwatch.elapsedMicroseconds / 1000} ms.', name: 'TfliteAiDataSource');

    return Future.value(imageData);
  }

  void dispose() {
    _chatbotInterpreter?.close();
    _textToImageInterpreter?.close();
    _imageModificationInterpreter?.close();
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created TfliteAiDataSource at {file_path} with hardware acceleration and profiling.")

# --- 6. Re-create chatbot_service.dart ---
file_path = os.path.join('lib', 'domain', 'usecases', 'chatbot_service.dart')
file_content = '''
import 'package:ai_chatbot_app/data/datasources/ai_data_source.dart';

// Abstract interface for chatbot interactions
abstract class ChatbotService {
  Future<String> sendMessage(String message);
}

// Concrete implementation of ChatbotService
class ChatbotServiceImpl implements ChatbotService {
  final AiDataSource _aiDataSource;

  ChatbotServiceImpl(this._aiDataSource);

  @override
  Future<String> sendMessage(String message) async {
    return await _aiDataSource.getChatbotResponse(message);
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ChatbotService at {file_path}")

# --- 7. Re-create chatbot_provider.dart ---
file_path = os.path.join('lib', 'presentation', 'providers', 'chatbot_provider.dart')
file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/domain/usecases/chatbot_service.dart';

class ChatMessage {
  final String text;
  final bool isUser;

  ChatMessage({required this.text, required this.isUser});
}

class ChatbotProvider with ChangeNotifier {
  final ChatbotService _chatbotService;
  final List<ChatMessage> _messages = [];
  bool _isLoading = false;

  ChatbotProvider(this._chatbotService);

  List<ChatMessage> get messages => _messages;
  bool get isLoading => _isLoading;

  Future<void> sendMessage(String text) async {
    if (text.trim().isEmpty) return;

    _messages.add(ChatMessage(text: text, isUser: true));
    _isLoading = true;
    notifyListeners();

    try {
      final response = await _chatbotService.sendMessage(text);
      _messages.add(ChatMessage(text: response, isUser: false));
    } catch (e) {
      _messages.add(ChatMessage(text: 'Error: Could not get a response. ($e)', isUser: false));
      print('ChatbotProvider error: $e');
    }

    _isLoading = false;
    notifyListeners();
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ChatbotProvider at {file_path}")

# --- 8. Re-create and fix chatbot_page.dart ---
file_path = os.path.join('lib', 'presentation', 'pages', 'chatbot_page.dart')
file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/presentation/providers/chatbot_provider.dart';

class ChatbotPage extends StatefulWidget {
  const ChatbotPage({super.key});

  @override
  State<ChatbotPage> createState() => _ChatbotPageState();
}

class _ChatbotPageState extends State<ChatbotPage> {
  final TextEditingController _textController = TextEditingController();
  final ScrollController _scrollController = ScrollController();

  @override
  void dispose() {
    _textController.dispose();
    _scrollController.dispose();
    super.dispose();
  }

  void _sendMessage() {
    if (_textController.text.isNotEmpty) {
      Provider.of<ChatbotProvider>(context, listen: false)
          .sendMessage(_textController.text);
      _textController.clear();
    }
  }

  void _scrollToBottom() {
    WidgetsBinding.instance.addPostFrameCallback((_) {
      if (_scrollController.hasClients) {
        _scrollController.animateTo(
          _scrollController.position.maxScrollExtent,
          duration: const Duration(milliseconds: 300),
          curve: Curves.easeOut,
        );
      }
    });
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        title: const Text('AI Chatbot'),
      ),
      body: Column(
        children: <Widget>[
          Expanded(
            child: Consumer<ChatbotProvider>(
              builder: (context, chatbotProvider, child) {
                if (chatbotProvider.messages.isNotEmpty && _scrollController.hasClients) {
                   _scrollToBottom();
                }
                return ListView.builder(
                  controller: _scrollController,
                  padding: const EdgeInsets.symmetric(horizontal: 10.0, vertical: 8.0),
                  itemCount: chatbotProvider.messages.length,
                  itemBuilder: (context, index) {
                    final message = chatbotProvider.messages[index];
                    final isUser = message.isUser;
                    return Padding(
                      padding: const EdgeInsets.symmetric(vertical: 4.0),
                      child: Row(
                        mainAxisAlignment: isUser
                            ? MainAxisAlignment.end
                            : MainAxisAlignment.start,
                        crossAxisAlignment: CrossAxisAlignment.end,
                        children: <Widget>[
                          if (!isUser) ...[
                            CircleAvatar(
                              backgroundColor: Theme.of(context).primaryColor,
                              child: const Icon(Icons.chat_bubble_outline, color: Colors.white, size: 20),
                            ),
                            const SizedBox(width: 8.0),
                          ],
                          Flexible(
                            child: Container(
                              constraints: BoxConstraints(
                                maxWidth: MediaQuery.of(context).size.width * 0.75,
                              ),
                              padding: const EdgeInsets.all(12.0),
                              decoration: BoxDecoration(
                                color: isUser
                                    ? Theme.of(context).colorScheme.primary
                                    : Theme.of(context).colorScheme.secondary.withOpacity(0.2),
                                borderRadius: BorderRadius.only(
                                  topLeft: const Radius.circular(15.0),
                                  topRight: const Radius.circular(15.0),
                                  bottomLeft: isUser
                                      ? const Radius.circular(15.0)
                                      : const Radius.circular(5.0),
                                  bottomRight: isUser
                                      ? const Radius.circular(5.0)
                                      : const Radius.circular(15.0),
                                ),
                              ),
                              child: Text(
                                message.text,
                                style: TextStyle(
                                  color: isUser ? Colors.white : Theme.of(context).textTheme.bodyLarge?.color,
                                ),
                              ),
                            ),
                          ),
                          if (isUser) ...[
                            const SizedBox(width: 8.0),
                            CircleAvatar(
                              backgroundColor: Theme.of(context).primaryColor,
                              child: const Icon(Icons.person, color: Colors.white, size: 20),
                            ),
                          ],
                        ],
                      ),
                    );
                  },
                );
              },
            ),
          ),
          if (Provider.of<ChatbotProvider>(context).isLoading)
            const Padding(
              padding: EdgeInsets.all(8.0),
              child: LinearProgressIndicator(),
            ),
          Padding(
            padding: const EdgeInsets.all(8.0),
            child: Row(
              children: <Widget>[
                Expanded(
                  child: TextField(
                    controller: _textController,
                    decoration: InputDecoration(
                      hintText: 'Enter your message...',
                    ).applyDefaults(Theme.of(context).inputDecorationTheme),
                    onSubmitted: (_) => _sendMessage(),
                  ),
                ),
                const SizedBox(width: 8.0),
                FloatingActionButton(
                  onPressed: _sendMessage,
                  backgroundColor: Theme.of(context).colorScheme.secondary,
                  foregroundColor: Theme.of(context).colorScheme.onSecondary,
                  child: const Icon(Icons.send),
                ),
              ],
            ),
          ),
        ],
      ),
    );
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ChatbotPage at {file_path} with enhanced message bubble aesthetics.")

# --- 9. Re-create image_generation_service.dart ---
file_path = os.path.join('lib', 'domain', 'usecases', 'image_generation_service.dart')
file_content = '''
import 'package:ai_chatbot_app/data/datasources/ai_data_source.dart';

// Abstract interface for text-to-image generation
abstract class ImageGenerationService {
  Future<List<int>> generateImage(String prompt);
}

// Concrete implementation of ImageGenerationService
class ImageGenerationServiceImpl implements ImageGenerationService {
  final AiDataSource _aiDataSource;

  ImageGenerationServiceImpl(this._aiDataSource);

  @override
  Future<List<int>> generateImage(String prompt) async {
    return await _aiDataSource.generateImageFromText(prompt);
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ImageGenerationService at {file_path}")

# --- 10. Re-create image_generation_provider.dart ---
file_path = os.path.join('lib', 'presentation', 'providers', 'image_generation_provider.dart')
file_content = '''
import 'package:flutter/material.dart';
import 'package:ai_chatbot_app/domain/usecases/image_generation_service.dart';

class ImageGenerationProvider with ChangeNotifier {
  final ImageGenerationService _imageGenerationService;
  List<int>? _generatedImageBytes;
  bool _isLoading = false;
  String? _errorMessage;

  ImageGenerationProvider(this._imageGenerationService);

  List<int>? get generatedImageBytes => _generatedImageBytes;
  bool get isLoading => _isLoading;
  String? get errorMessage => _errorMessage;

  Future<void> generateImage(String prompt) async {
    if (prompt.trim().isEmpty) {
      _errorMessage = 'Please enter a prompt to generate an image.';
      notifyListeners();
      return;
    }

    _isLoading = true;
    _errorMessage = null;
    _generatedImageBytes = null;
    notifyListeners();

    try {
      _generatedImageBytes = await _imageGenerationService.generateImage(prompt);
    } catch (e) {
      _errorMessage = 'Error generating image: $e';
      print('ImageGenerationProvider error: $e');
    }

    _isLoading = false;
    notifyListeners();
  }

  void clearImage() {
    _generatedImageBytes = null;
    _errorMessage = null;
    notifyListeners();
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ImageGenerationProvider at {file_path}")

# --- 11. Re-create and fix image_generation_page.dart ---
file_path = os.path.join('lib', 'presentation', 'pages', 'image_generation_page.dart')
file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/presentation/providers/image_generation_provider.dart';
import 'dart:typed_data'; // Required for Uint8List.fromList

class ImageGenerationPage extends StatefulWidget {
  const ImageGenerationPage({super.key});

  @override
  State<ImageGenerationPage> createState() => _ImageGenerationPageState();
}

class _ImageGenerationPageState extends State<ImageGenerationPage> {
  final TextEditingController _promptController = TextEditingController();

  @override
  void dispose() {
    _promptController.dispose();
    super.dispose();
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        title: const Text('Text-to-Image Generator'),
      ),
      body: Consumer<ImageGenerationProvider>(
        builder: (context, imageProvider, child) {
          return SingleChildScrollView(
            padding: const EdgeInsets.all(16.0),
            child: Column(
              crossAxisAlignment: CrossAxisAlignment.stretch,
              children: <Widget>[
                TextField(
                  controller: _promptController,
                  decoration: InputDecoration(
                    hintText: 'Enter your image prompt (e.g., "a cat in space")',
                  ).applyDefaults(Theme.of(context).inputDecorationTheme),
                  maxLines: 3,
                ),
                const SizedBox(height: 16.0),

                imageProvider.isLoading
                    ? Center(
                        child: Column(
                          children: [
                            const CircularProgressIndicator(),
                            const SizedBox(height: 8.0),
                            Text('Generating image...', style: Theme.of(context).textTheme.bodyMedium),
                          ],
                        ),
                      )
                    : ElevatedButton(
                        onPressed: () {
                          imageProvider.generateImage(_promptController.text);
                        },
                        child: const Text('Generate Image'),
                      ),
                const SizedBox(height: 16.0),

                if (imageProvider.errorMessage != null)
                  Padding(
                    padding: const EdgeInsets.only(bottom: 16.0),
                    child: Text(
                      imageProvider.errorMessage!,
                      style: Theme.of(context).textTheme.bodyMedium?.copyWith(color: Colors.red),
                      textAlign: TextAlign.center,
                    ),
                  ),

                if (imageProvider.generatedImageBytes != null)
                  Column(
                    children: [
                      const Text('Generated Image:', style: TextStyle(fontWeight: FontWeight.bold)),
                      const SizedBox(height: 8.0),
                      Container(
                        decoration: BoxDecoration(
                          border: Border.all(color: Theme.of(context).colorScheme.primary.withOpacity(0.5), width: 2),
                          borderRadius: BorderRadius.circular(10),
                        ),
                        child: ClipRRect(
                          borderRadius: BorderRadius.circular(8),
                          child: Image.memory(
                            Uint8List.fromList(imageProvider.generatedImageBytes!),
                            fit: BoxFit.contain,
                            errorBuilder: (context, error, stackTrace) => const Icon(Icons.error_outline, color: Colors.red, size: 50),
                          ),
                        ),
                      ),
                      const SizedBox(height: 16.0),
                      ElevatedButton.icon(
                        onPressed: imageProvider.clearImage,
                        icon: const Icon(Icons.clear),
                        label: const Text('Clear Image'),
                      ),
                    ],
                  )
                else if (!imageProvider.isLoading)
                  Padding(
                    padding: const EdgeInsets.all(24.0),
                    child: Text(
                      """No image generated yet.\nEnter a prompt and click "Generate Image""",
                      textAlign: TextAlign.center,
                      style: Theme.of(context).textTheme.bodyMedium?.copyWith(color: Colors.grey[600]),
                    ),
                  ),
              ],
            ),
          );
        },
      ),
    );
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ImageGenerationPage at {file_path} with improved UI and visual cues.")

# --- 12. Re-create image_modification_service.dart ---
file_path = os.path.join('lib', 'domain', 'usecases', 'image_modification_service.dart')
file_content = '''
import 'package:ai_chatbot_app/data/datasources/ai_data_source.dart';

// Abstract interface for image modification
abstract class ImageModificationService {
  Future<List<int>> modifyImage(List<int> imageData, String prompt);
}

// Concrete implementation of ImageModificationService
class ImageModificationServiceImpl implements ImageModificationService {
  final AiDataSource _aiDataSource;

  ImageModificationServiceImpl(this._aiDataSource);

  @override
  Future<List<int>> modifyImage(List<int> imageData, String prompt) async {
    return await _aiDataSource.modifyImage(imageData, prompt);
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ImageModificationService at {file_path}")

# --- 13. Re-create image_modification_provider.dart ---
file_path = os.path.join('lib', 'presentation', 'providers', 'image_modification_provider.dart')
file_content = '''
import 'package:flutter/material.dart';
import 'package:image_picker/image_picker.dart';
import 'dart:io';
import 'package:ai_chatbot_app/domain/usecases/image_modification_service.dart';

class ImageModificationProvider with ChangeNotifier {
  final ImageModificationService _imageModificationService;

  File? _originalImageFile;
  List<int>? _originalImageBytes;
  String _prompt = '';
  List<int>? _modifiedImageBytes;
  bool _isLoading = false;
  String? _errorMessage;

  ImageModificationProvider(this._imageModificationService);

  File? get originalImageFile => _originalImageFile;
  List<int>? get originalImageBytes => _originalImageBytes;
  String get prompt => _prompt;
  List<int>? get modifiedImageBytes => _modifiedImageBytes;
  bool get isLoading => _isLoading;
  String? get errorMessage => _errorMessage;

  void setPrompt(String newPrompt) {
    _prompt = newPrompt;
    notifyListeners();
  }

  Future<void> pickImage() async {
    final ImagePicker picker = ImagePicker();
    final XFile? image = await picker.pickImage(source: ImageSource.gallery);

    if (image != null) {
      _originalImageFile = File(image.path);
      _originalImageBytes = await _originalImageFile!.readAsBytes();
      _modifiedImageBytes = null;
      _errorMessage = null;
      notifyListeners();
    }
  }

  Future<void> modifyImage() async {
    if (_originalImageBytes == null) {
      _errorMessage = 'Please pick an image first.';
      notifyListeners();
      return;
    }
    if (_prompt.trim().isEmpty) {
      _errorMessage = 'Please enter a prompt for modification.';
      notifyListeners();
      return;
    }

    _isLoading = true;
    _errorMessage = null;
    _modifiedImageBytes = null;
    notifyListeners();

    try {
      _modifiedImageBytes = await _imageModificationService.modifyImage(
          _originalImageBytes!, _prompt);
    } catch (e) {
      _errorMessage = 'Error modifying image: $e';
      print('ImageModificationProvider error: $e');
    }

    _isLoading = false;
    notifyListeners();
  }

  void clearAll() {
    _originalImageFile = null;
    _originalImageBytes = null;
    _prompt = '';
    _modifiedImageBytes = null;
    _errorMessage = null;
    notifyListeners();
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ImageModificationProvider at {file_path}")

# --- 14. Re-create and fix image_modification_page.dart ---
file_path = os.path.join('lib', 'presentation', 'pages', 'image_modification_page.dart')
file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/presentation/providers/image_modification_provider.dart';
import 'dart:typed_data'; // Required for Uint8List.fromList

class ImageModificationPage extends StatefulWidget {
  const ImageModificationPage({super.key});

  @override
  State<ImageModificationPage> createState() => _ImageModificationPageState();
}

class _ImageModificationPageState extends State<ImageModificationPage> {
  final TextEditingController _promptController = TextEditingController();

  @override
  void initState() {
    super.initState();
    _promptController.addListener(() {
      Provider.of<ImageModificationProvider>(context, listen: false)
          .setPrompt(_promptController.text);
    });
  }

  @override
  void dispose() {
    _promptController.dispose();
    super.dispose();
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        title: const Text('Image Modifier'),
        actions: [
          IconButton(
            icon: const Icon(Icons.clear_all),
            onPressed: () {
              Provider.of<ImageModificationProvider>(context, listen: false).clearAll();
              _promptController.clear();
            },
          ),
        ],
      ),
      body: Consumer<ImageModificationProvider>(
        builder: (context, imageModifierProvider, child) {
          return SingleChildScrollView(
            padding: const EdgeInsets.all(16.0),
            child: Column(
              crossAxisAlignment: CrossAxisAlignment.stretch,
              children: <Widget>[
                ElevatedButton.icon(
                  onPressed: imageModifierProvider.pickImage,
                  icon: const Icon(Icons.image),
                  label: const Text('Pick Image to Modify'),
                  style: Theme.of(context).elevatedButtonTheme.style?.copyWith(
                    backgroundColor: MaterialStateProperty.all(Theme.of(context).primaryColor),
                    foregroundColor: MaterialStateProperty.all(Colors.white),
                  ),
                ),
                const SizedBox(height: 16.0),

                TextField(
                  controller: _promptController,
                  decoration: InputDecoration(
                    hintText: 'Enter modification prompt (e.g., "make it vintage", "add sunglasses")',
                  ).applyDefaults(Theme.of(context).inputDecorationTheme),
                  maxLines: 3,
                ),
                const SizedBox(height: 16.0),

                imageModifierProvider.isLoading
                    ? Center(
                        child: Column(
                          children: [
                            const CircularProgressIndicator(),
                            const SizedBox(height: 8.0),
                            Text('Modifying image...', style: Theme.of(context).textTheme.bodyMedium),
                          ],
                        ),
                      )
                    : ElevatedButton.icon(
                        onPressed: imageModifierProvider.originalImageBytes != null && _promptController.text.isNotEmpty
                            ? imageModifierProvider.modifyImage
                            : null,
                        icon: const Icon(Icons.auto_fix_high),
                        label: const Text('Apply Modification'),
                      ),
                const SizedBox(height: 16.0),

                if (imageModifierProvider.errorMessage != null)
                  Padding(
                    padding: const EdgeInsets.only(bottom: 16.0),
                    child: Text(
                      imageModifierProvider.errorMessage!,
                      style: Theme.of(context).textTheme.bodyMedium?.copyWith(color: Colors.red),
                      textAlign: TextAlign.center,
                    ),
                  ),

                if (imageModifierProvider.originalImageBytes != null)
                  Column(
                    crossAxisAlignment: CrossAxisAlignment.stretch,
                    children: [
                      Text('Original Image:', style: Theme.of(context).textTheme.headlineSmall?.copyWith(fontWeight: FontWeight.bold)),
                      const SizedBox(height: 8.0),
                      Container(
                        decoration: BoxDecoration(
                          border: Border.all(color: Theme.of(context).colorScheme.primary.withOpacity(0.5), width: 2),
                          borderRadius: BorderRadius.circular(10),
                        ),
                        child: ClipRRect(
                          borderRadius: BorderRadius.circular(8),
                          child: Image.memory(Uint8List.fromList(imageModifierProvider.originalImageBytes!),
                              height: 200, fit: BoxFit.contain),
                        ),
                      ),
                      const SizedBox(height: 16.0),
                    ],
                  ),

                if (imageModifierProvider.modifiedImageBytes != null)
                  Column(
                    crossAxisAlignment: CrossAxisAlignment.stretch,
                    children: [
                      Text('Modified Image:', style: Theme.of(context).textTheme.headlineSmall?.copyWith(fontWeight: FontWeight.bold)),
                      const SizedBox(height: 8.0),
                      Container(
                        decoration: BoxDecoration(
                          border: Border.all(color: Theme.of(context).colorScheme.secondary.withOpacity(0.5), width: 2),
                          borderRadius: BorderRadius.circular(10),
                        ),
                        child: ClipRRect(
                          borderRadius: BorderRadius.circular(8),
                          child: Image.memory(Uint8List.fromList(imageModifierProvider.modifiedImageBytes!),
                              height: 200, fit: BoxFit.contain),
                        ),
                      ),
                      const SizedBox(height: 16.0),
                    ],
                  ),
              ],
            ),
          );
        },
      ),
    );
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ImageModificationPage at {file_path} with improved UI and visual cues.")

# --- 15. Re-create main.dart (including theme and navigation) ---
file_path = os.path.join('lib', 'main.dart')
file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/data/datasources/ai_data_source.dart';
import 'package:ai_chatbot_app/data/datasources/tflite_ai_data_source.dart';
import 'package:ai_chatbot_app/domain/usecases/chatbot_service.dart';
import 'package:ai_chatbot_app/domain/usecases/image_generation_service.dart';
import 'package:ai_chatbot_app/domain/usecases/image_modification_service.dart';
import 'package:ai_chatbot_app/presentation/pages/chatbot_page.dart';
import 'package:ai_chatbot_app/presentation/pages/image_generation_page.dart';
import 'package:ai_chatbot_app/presentation/pages/image_modification_page.dart';
import 'package:ai_chatbot_app/presentation/providers/chatbot_provider.dart';
import 'package:ai_chatbot_app/presentation/providers/image_generation_provider.dart';
import 'package:ai_chatbot_app/presentation/providers/image_modification_provider.dart';

void main() async {
  WidgetsFlutterBinding.ensureInitialized();

  final AiDataSource aiDataSource = TfliteAiDataSource();
  await (aiDataSource as TfliteAiDataSource).init(); // Initialize models

  final ChatbotService chatbotService = ChatbotServiceImpl(aiDataSource);
  final ImageGenerationService imageGenerationService = ImageGenerationServiceImpl(aiDataSource);
  final ImageModificationService imageModificationService = ImageModificationServiceImpl(aiDataSource);

  runApp(
    MultiProvider(
      providers: [
        ChangeNotifierProvider(
          create: (_) => ChatbotProvider(chatbotService),
        ),
        ChangeNotifierProvider(
          create: (_) => ImageGenerationProvider(imageGenerationService),
        ),
        ChangeNotifierProvider(
          create: (_) => ImageModificationProvider(imageModificationService),
        ),
      ],
      child: const MyApp(),
    ),
  );
}

class MyApp extends StatelessWidget {
  const MyApp({super.key});

  @override
  Widget build(BuildContext context) {
    return MaterialApp(
      title: 'AI Chatbot App',
      theme: ThemeData(
        primarySwatch: Colors.blueGrey,
        primaryColor: Colors.blueGrey[700],
        colorScheme: ColorScheme.fromSwatch(
          primarySwatch: Colors.blueGrey,
        ).copyWith(
          secondary: Colors.tealAccent[400],
        ),
        scaffoldBackgroundColor: Colors.grey[50],
        appBarTheme: AppBarTheme(
          backgroundColor: Colors.blueGrey[800],
          foregroundColor: Colors.white,
          elevation: 0,
          centerTitle: true,
          titleTextStyle: TextStyle(
            color: Colors.white,
            fontSize: 20,
            fontWeight: FontWeight.bold,
          ),
        ),
        elevatedButtonTheme: ElevatedButtonThemeData(
          style: ElevatedButton.styleFrom(
            backgroundColor: Colors.tealAccent[400],
            foregroundColor: Colors.blueGrey[900],
            padding: EdgeInsets.symmetric(horizontal: 20, vertical: 15),
            shape: RoundedRectangleBorder(
              borderRadius: BorderRadius.circular(10),
            ),
          ),
        ),
        inputDecorationTheme: InputDecorationTheme(
          border: OutlineInputBorder(
            borderRadius: BorderRadius.circular(10),
            borderSide: BorderSide.none,
          ),
          filled: true,
          fillColor: Colors.grey[200],
          contentPadding: EdgeInsets.symmetric(horizontal: 16, vertical: 12),
          hintStyle: TextStyle(color: Colors.grey[600]),
        ),
        textTheme: TextTheme(
          headlineMedium: TextStyle(color: Colors.blueGrey[800], fontWeight: FontWeight.bold),
          bodyLarge: TextStyle(color: Colors.blueGrey[700]),
          bodyMedium: TextStyle(color: Colors.blueGrey[600]),
          labelLarge: TextStyle(color: Colors.blueGrey[900], fontWeight: FontWeight.bold),
        ),
        visualDensity: VisualDensity.adaptivePlatformDensity,
      ),
      home: const MainScreen(),
    );
  }
}

class MainScreen extends StatefulWidget {
  const MainScreen({super.key});

  @override
  State<MainScreen> createState() => _MainScreenState();
}

class _MainScreenState extends State<MainScreen> {
  int _selectedIndex = 0;

  static const List<Widget> _widgetOptions = <Widget>[
    ChatbotPage(),
    ImageGenerationPage(),
    ImageModificationPage(),
  ];

  void _onItemTapped(int index) {
    setState(() {
      _selectedIndex = index;
    });
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      body: Center(
        child: _widgetOptions.elementAt(_selectedIndex),
      ),
      bottomNavigationBar: BottomNavigationBar(
        items: const <BottomNavigationBarItem>[
          BottomNavigationBarItem(
            icon: Icon(Icons.chat),
            label: 'Chatbot',
          ),
          BottomNavigationBarItem(
            icon: Icon(Icons.image),
            label: 'Image Gen',
          ),
          BottomNavigationBarItem(
            icon: Icon(Icons.edit),
            label: 'Image Modify',
          ),
        ],
        currentIndex: _selectedIndex,
        selectedItemColor: Colors.tealAccent[400],
        unselectedItemColor: Colors.blueGrey[400],
        backgroundColor: Colors.blueGrey[800],
        onTap: _onItemTapped,
      ),
    );
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created main.dart to integrate all features and theme at {file_path}")

# --- 16. Update pubspec.yaml with dependencies ---
pubspec_path = 'pubspec.yaml'
with open(pubspec_path, 'r') as f:
    pubspec_content = f.readlines()

updated_content = []
inserted_deps = False
for line in pubspec_content:
    updated_content.append(line)
    if 'sdk: flutter' in line and not inserted_deps:
        updated_content.append(
            "  tflite_flutter: ^0.12.1\n"
            "  provider: ^6.1.1\n"
            "  image_picker: ^1.0.7\n"
            "  image: ^4.1.7\n"
            "  path_provider: ^2.1.2\n"
        )
        inserted_deps = True

# Also ensure assets/models/ is declared
assets_declared = False
flutter_section_found = False
for i, line in enumerate(updated_content):
    if line.strip() == 'flutter:':
        flutter_section_found = True
        # Check if assets section exists below flutter: and add if not
        for j in range(i + 1, len(updated_content)):
            if updated_content[j].strip().startswith('assets:'):
                assets_declared = True
                break
        if not assets_declared:
            updated_content.insert(i + 1, '  assets:\n') # Insert assets block
            updated_content.insert(i + 2, '    - assets/models/\n')
            assets_declared = True
            break

    if flutter_section_found and '    - assets/models/' in line:
        assets_declared = True
        break

if not assets_declared and flutter_section_found:
    # If 'flutter:' exists but no 'assets:' section was found and modified above, add it at the end of the flutter block
    # Find last line of flutter block
    last_flutter_line_idx = -1
    for i in range(len(updated_content) -1, -1, -1):
        if updated_content[i].strip() != '' and (updated_content[i].startswith(' ') and not updated_content[i].startswith('  -')):
            last_flutter_line_idx = i + 1
            break
    if last_flutter_line_idx == -1:
        last_flutter_line_idx = len(updated_content)

    updated_content.insert(last_flutter_line_idx, '  assets:\n')
    updated_content.insert(last_flutter_line_idx + 1, '    - assets/models/\n')

with open(pubspec_path, 'w') as f:
    f.writelines(updated_content)
print("Dependencies and assets re-added to pubspec.yaml.")

# --- 17. Run flutter pub get ---
print("Running flutter pub get...")
!flutter pub get
print("flutter pub get completed.")

# --- 18. Fix build.gradle.kts to specify SDK versions ---
# The build.gradle.kts might not be generated in every project, so check existence first.
build_gradle_kts_path = os.path.join('android', 'app', 'build.gradle.kts')

if os.path.exists(build_gradle_kts_path):
    print(f"Modifying {build_gradle_kts_path} to set compileSdkVersion, targetSdkVersion, and buildToolsVersion to 33...")
    with open(build_gradle_kts_path, 'r') as f:
        content = f.readlines()

    updated_content_gradle = []
    for line in content:
        if 'compileSdkVersion' in line:
            updated_content_gradle.append("    compileSdkVersion = 33\n")
        elif 'buildToolsVersion' in line:
            updated_content_gradle.append('    buildToolsVersion = "33.0.0"\n')
        elif 'targetSdk' in line and 'flutter.targetSdkVersion' in line:
            updated_content_gradle.append("        targetSdk = 33\n")
        else:
            updated_content_gradle.append(line)

    with open(build_gradle_kts_path, 'w') as f:
        f.writelines(updated_content_gradle)
    print(f"Updated {build_gradle_kts_path} with compileSdkVersion = 33, targetSdk = 33, and buildToolsVersion = \"33.0.0\".")
else:
    print(f"Warning: {build_gradle_kts_path} not found. Skipping build.gradle.kts modification.")

# --- 19. Run flutter clean ---
print("Running flutter clean...")
!flutter clean
print("flutter clean completed.")

# --- 20. Run flutter doctor -v to verify Android SDK setup ---
print("Verifying Android SDK setup with flutter doctor -v...")
!flutter doctor -v

# --- 21. Run flutter build apk --analyze-size --target-platform=android-arm64 ---
print("Running flutter build apk --analyze-size --target-platform=android-arm64...")
!flutter build apk --analyze-size --target-platform=android-arm64

# Navigate back to the original directory
os.chdir('..')
print(f"Navigated back to: {os.getcwd()}")
````

````python
import os
import subprocess
import shutil

project_root = 'ai_chatbot_app'

# --- 0. Ensure Flutter SDK is installed and in PATH for this session ---
print("Ensuring Flutter SDK is installed and in PATH...")
flutter_install_path = '/usr/local/flutter'
if not os.path.exists(flutter_install_path):
    print("Cloning Flutter SDK...")
    !git clone https://github.com/flutter/flutter.git {flutter_install_path} > /dev/null
    print("Flutter SDK cloned.")

# Add Flutter to PATH if not already there
if f"{flutter_install_path}/bin" not in os.environ["PATH"]:
    os.environ["PATH"] += os.pathsep + f"{flutter_install_path}/bin"
    print("Flutter SDK added to PATH.")

# --- 1. Clean and Re-initialize Flutter project ---
if os.path.exists(project_root):
    print(f"Removing existing project directory: {project_root}")
    shutil.rmtree(project_root)

print("Initializing a new Flutter project named 'ai_chatbot_app'...")
# Ensure flutter create uses --no-overwrite to prevent interactive prompts if run multiple times
!flutter create --no-overwrite {project_root}
print("Flutter project 'ai_chatbot_app' initialized successfully.")

# Navigate to the project directory
os.chdir(project_root)
print(f"Navigated to: {os.getcwd()}")

# --- 2. Re-install Android SDK & Configure (as previously done) ---
print("Re-configuring Android SDK environment...")
ANDROID_SDK_ROOT = '/opt/android-sdk'
cmdline_tools_url = 'https://dl.google.com/android/repository/commandlinetools-linux-10406996_latest.zip'

# Create necessary directories
os.makedirs(ANDROID_SDK_ROOT, exist_ok=True)
os.makedirs(os.path.join(ANDROID_SDK_ROOT, 'cmdline-tools', 'latest'), exist_ok=True)

# Download command line tools using wget
zip_path = '/tmp/commandlinetools.zip'
unzip_dir = '/tmp/cmdline-tools-unzipped'

print("Downloading Android SDK command-line tools...")
return_code = subprocess.call(['wget', '-q', '-O', zip_path, cmdline_tools_url])

if return_code != 0:
    print(f"Error: Failed to download command-line tools. Wget returned code {return_code}.")
else:
    print("Download successful. Unzipping and moving files...")
    os.makedirs(unzip_dir, exist_ok=True)
    !unzip -q {zip_path} -d {unzip_dir}
    source_path = os.path.join(unzip_dir, 'cmdline-tools')
    target_path = os.path.join(ANDROID_SDK_ROOT, 'cmdline-tools', 'latest')
    if os.path.exists(source_path):
        !mv {source_path}/* {target_path}/
        print("Files moved.")
    else:
        print(f"Warning: '{source_path}' not found after unzip. Skipping move.")

# Set ANDROID_HOME and add tools to PATH
os.environ['ANDROID_HOME'] = ANDROID_SDK_ROOT
os.environ['PATH'] += os.pathsep + os.path.join(ANDROID_SDK_ROOT, 'cmdline-tools', 'latest', 'bin')
os.environ['PATH'] += os.pathsep + os.path.join(ANDROID_SDK_ROOT, 'platform-tools')

# Install necessary SDK components (platforms, build-tools)
sdkmanager_path = os.path.join(ANDROID_SDK_ROOT, 'cmdline-tools', 'latest', 'bin', 'sdkmanager')
print("Installing Android SDK components...")
!yes | "{sdkmanager_path}" --install "platform-tools" "platforms;android-33" "build-tools;33.0.0" "platforms;android-36" "build-tools;28.0.3" > /dev/null

# Accept licenses
print("Accepting Android licenses...")
!yes | "{sdkmanager_path}" --licenses > /dev/null

# Explicitly set Flutter's Android SDK path
print(f"Setting Flutter's Android SDK path to {os.environ['ANDROID_HOME']}...")
!flutter config --android-sdk="{os.environ['ANDROID_HOME']}"

print("Android SDK re-configured.")

# --- 3. Re-create Initial Folder Structure ---
lib_path = 'lib' # Relative to current working directory

directories_to_create = [
    'data',
    'data/datasources',
    'data/models',
    'data/repositories',
    'domain',
    'domain/entities',
    'domain/repositories',
    'domain/usecases',
    'presentation',
    'presentation/pages',
    'presentation/widgets',
    'presentation/providers',
    'core',
    'core/utils',
    'core/constants'
]

for directory in directories_to_create:
    path = os.path.join(lib_path, directory)
    os.makedirs(path, exist_ok=True)
print("Initial folder structure for clean architecture re-established.")

# --- 4. Re-create ai_data_source.dart ---
file_path = os.path.join('lib', 'data', 'datasources', 'ai_data_source.dart')
file_content = '''
abstract class AiDataSource {
  // Chatbot functionalities
  Future<String> getChatbotResponse(String query);

  // Text-to-Image generation functionalities
  Future<List<int>> generateImageFromText(String prompt);

  // Image modification functionalities
  Future<List<int>> modifyImage(List<int> imageData, String prompt);
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created abstract class AiDataSource at {file_path}")

# --- 5. Re-create and fix tflite_ai_data_source.dart ---
file_path = os.path.join('lib', 'data', 'datasources', 'tflite_ai_data_source.dart')
file_content = '''
import 'package:ai_chatbot_app/data/datasources/ai_data_source.dart';
import 'package:flutter/services.dart' show rootBundle; // For loading assets
import 'package:tflite_flutter/tflite_flutter.dart' as tfl; // Use alias for tflite_flutter
import 'package:image/image.dart' as img; // For image manipulation
import 'dart:typed_data'; // For ByteData
import 'dart:developer' as developer; // For logging performance

class TfliteAiDataSource implements AiDataSource {
  tfl.Interpreter? _chatbotInterpreter;
  tfl.Interpreter? _textToImageInterpreter;
  tfl.Interpreter? _imageModificationInterpreter;

  // Models will be loaded asynchronously during initialization
  TfliteAiDataSource();

  Future<void> init() async {
    tfl.InterpreterOptions options = tfl.InterpreterOptions();

    // Attempt to add GPU delegate first
    try {
      options.addDelegate(tfl.GpuDelegateV2());
      print('GPU Delegate loaded.');
    } catch (e) {
      print('Failed to load GPU Delegate: $e');
      // Fallback to NNAPI if GPU fails
      try {
        options.addDelegate(tfl.NnApiDelegate());
        print('NNAPI Delegate loaded.');
      } catch (e) {
        print('Failed to load NNAPI Delegate: $e');
        print('No hardware acceleration delegate loaded. Falling back to CPU.');
      }
    }

    // Load chatbot model
    try {
      _chatbotInterpreter = await tfl.Interpreter.fromAsset('assets/models/chatbot_nlu_model.tflite', options: options);
      print('Chatbot NLU model loaded successfully.');
    } catch (e) {
      print('Failed to load chatbot NLU model: $e');
    }

    // Load text-to-image model (placeholder)
    try {
      _textToImageInterpreter = await tfl.Interpreter.fromAsset('assets/models/text_to_image_model.tflite', options: options);
      print('Text-to-Image model loaded successfully.');
    } catch (e) {
      print('Failed to load text-to-image model: $e');
    }

    // Load image modification model (placeholder)
    try {
      _imageModificationInterpreter = await tfl.Interpreter.fromAsset('assets/models/image_modification_model.tflite', options: options);
      print('Image Modification model loaded successfully.');
    } catch (e) {
      print('Failed to load image modification model: $e');
    }
  }

  @override
  Future<String> getChatbotResponse(String query) async {
    if (_chatbotInterpreter == null) {
      throw Exception('Chatbot model not loaded.');
    }

    final stopwatch = Stopwatch()..start();

    developer.log('Chatbot preprocessing started.', name: 'TfliteAiDataSource');
    final preprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 10));
    final preprocessingTime = stopwatch.elapsedMicroseconds - preprocessingStart;
    developer.log('Chatbot preprocessing finished in ${preprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Chatbot inference started.', name: 'TfliteAiDataSource');
    final inferenceStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 50));
    final inferenceTime = stopwatch.elapsedMicroseconds - inferenceStart;
    developer.log('Chatbot inference finished in ${inferenceTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Chatbot post-processing started.', name: 'TfliteAiDataSource');
    final postprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 5));
    final postprocessingTime = stopwatch.elapsedMicroseconds - postprocessingStart;
    developer.log('Chatbot post-processing finished in ${postprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    stopwatch.stop();
    developer.log('Total Chatbot response time for \'$query\': ${stopwatch.elapsedMicroseconds / 1000} ms.', name: 'TfliteAiDataSource');

    return Future.value('This is a placeholder response for: $query. (NLU processing to be integrated)');
  }

  @override
  Future<List<int>> generateImageFromText(String prompt) async {
    if (_textToImageInterpreter == null) {
      throw Exception('Text-to-Image model not loaded.');
    }

    final stopwatch = Stopwatch()..start();

    developer.log('Text-to-Image preprocessing started.', name: 'TfliteAiDataSource');
    final preprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 20));
    final preprocessingTime = stopwatch.elapsedMicroseconds - preprocessingStart;
    developer.log('Text-to-Image preprocessing finished in ${preprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Text-to-Image inference started.', name: 'TfliteAiDataSource');
    final inferenceStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 500));
    final inferenceTime = stopwatch.elapsedMicroseconds - inferenceStart;
    developer.log('Text-to-Image inference finished in ${inferenceTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Text-to-Image post-processing started.', name: 'TfliteAiDataSource');
    final postprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 50));
    final postprocessingTime = stopwatch.elapsedMicroseconds - postprocessingStart;
    developer.log('Text-to-Image post-processing finished in ${postprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    stopwatch.stop();
    developer.log('Total Text-to-Image generation time for \'$prompt\': ${stopwatch.elapsedMicroseconds / 1000} ms.', name: 'TfliteAiDataSource');

    final dummyImage = img.Image(width: 256, height: 256);
    img.fill(dummyImage, color: img.ColorRgba8(0, 0, 0, 0));
    return img.encodePng(dummyImage);
  }

  @override
  Future<List<int>> modifyImage(List<int> imageData, String prompt) async {
    if (_imageModificationInterpreter == null) {
      throw Exception('Image Modification model not loaded.');
    }

    final stopwatch = Stopwatch()..start();

    developer.log('Image Modification preprocessing started.', name: 'TfliteAiDataSource');
    final preprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 30));
    final preprocessingTime = stopwatch.elapsedMicroseconds - preprocessingStart;
    developer.log('Image Modification preprocessing finished in ${preprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Image Modification inference started.', name: 'TfliteAiDataSource');
    final inferenceStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 700));
    final inferenceTime = stopwatch.elapsedMicroseconds - inferenceStart;
    developer.log('Image Modification inference finished in ${inferenceTime / 1000} ms.', name: 'TfliteAiDataSource');

    developer.log('Image Modification post-processing started.', name: 'TfliteAiDataSource');
    final postprocessingStart = stopwatch.elapsedMicroseconds;
    await Future.delayed(const Duration(milliseconds: 60));
    final postprocessingTime = stopwatch.elapsedMicroseconds - postprocessingStart;
    developer.log('Image Modification post-processing finished in ${postprocessingTime / 1000} ms.', name: 'TfliteAiDataSource');

    stopwatch.stop();
    developer.log('Total Image Modification time for \'$prompt\': ${stopwatch.elapsedMicroseconds / 1000} ms.', name: 'TfliteAiDataSource');

    return Future.value(imageData);
  }

  void dispose() {
    _chatbotInterpreter?.close();
    _textToImageInterpreter?.close();
    _imageModificationInterpreter?.close();
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created TfliteAiDataSource at {file_path} with hardware acceleration and profiling.")

# --- 6. Re-create chatbot_service.dart ---
file_path = os.path.join('lib', 'domain', 'usecases', 'chatbot_service.dart')
file_content = '''
import 'package:ai_chatbot_app/data/datasources/ai_data_source.dart';

// Abstract interface for chatbot interactions
abstract class ChatbotService {
  Future<String> sendMessage(String message);
}

// Concrete implementation of ChatbotService
class ChatbotServiceImpl implements ChatbotService {
  final AiDataSource _aiDataSource;

  ChatbotServiceImpl(this._aiDataSource);

  @override
  Future<String> sendMessage(String message) async {
    return await _aiDataSource.getChatbotResponse(message);
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ChatbotService at {file_path}")

# --- 7. Re-create chatbot_provider.dart ---
file_path = os.path.join('lib', 'presentation', 'providers', 'chatbot_provider.dart')
file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/domain/usecases/chatbot_service.dart';

class ChatMessage {
  final String text;
  final bool isUser;

  ChatMessage({required this.text, required this.isUser});
}

class ChatbotProvider with ChangeNotifier {
  final ChatbotService _chatbotService;
  final List<ChatMessage> _messages = [];
  bool _isLoading = false;

  ChatbotProvider(this._chatbotService);

  List<ChatMessage> get messages => _messages;
  bool get isLoading => _isLoading;

  Future<void> sendMessage(String text) async {
    if (text.trim().isEmpty) return;

    _messages.add(ChatMessage(text: text, isUser: true));
    _isLoading = true;
    notifyListeners();

    try {
      final response = await _chatbotService.sendMessage(text);
      _messages.add(ChatMessage(text: response, isUser: false));
    } catch (e) {
      _messages.add(ChatMessage(text: 'Error: Could not get a response. ($e)', isUser: false));
      print('ChatbotProvider error: $e');
    }

    _isLoading = false;
    notifyListeners();
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ChatbotProvider at {file_path}")

# --- 8. Re-create and fix chatbot_page.dart ---
file_path = os.path.join('lib', 'presentation', 'pages', 'chatbot_page.dart')
file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/presentation/providers/chatbot_provider.dart';

class ChatbotPage extends StatefulWidget {
  const ChatbotPage({super.key});

  @override
  State<ChatbotPage> createState() => _ChatbotPageState();
}

class _ChatbotPageState extends State<ChatbotPage> {
  final TextEditingController _textController = TextEditingController();
  final ScrollController _scrollController = ScrollController();

  @override
  void dispose() {
    _textController.dispose();
    _scrollController.dispose();
    super.dispose();
  }

  void _sendMessage() {
    if (_textController.text.isNotEmpty) {
      Provider.of<ChatbotProvider>(context, listen: false)
          .sendMessage(_textController.text);
      _textController.clear();
    }
  }

  void _scrollToBottom() {
    WidgetsBinding.instance.addPostFrameCallback((_) {
      if (_scrollController.hasClients) {
        _scrollController.animateTo(
          _scrollController.position.maxScrollExtent,
          duration: const Duration(milliseconds: 300),
          curve: Curves.easeOut,
        );
      }
    });
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        title: const Text('AI Chatbot'),
      ),
      body: Column(
        children: <Widget>[
          Expanded(
            child: Consumer<ChatbotProvider>(
              builder: (context, chatbotProvider, child) {
                if (chatbotProvider.messages.isNotEmpty && _scrollController.hasClients) {
                   _scrollToBottom();
                }
                return ListView.builder(
                  controller: _scrollController,
                  padding: const EdgeInsets.symmetric(horizontal: 10.0, vertical: 8.0),
                  itemCount: chatbotProvider.messages.length,
                  itemBuilder: (context, index) {
                    final message = chatbotProvider.messages[index];
                    final isUser = message.isUser;
                    return Padding(
                      padding: const EdgeInsets.symmetric(vertical: 4.0),
                      child: Row(
                        mainAxisAlignment: isUser
                            ? MainAxisAlignment.end
                            : MainAxisAlignment.start,
                        crossAxisAlignment: CrossAxisAlignment.end,
                        children: <Widget>[
                          if (!isUser) ...[
                            CircleAvatar(
                              backgroundColor: Theme.of(context).primaryColor,
                              child: const Icon(Icons.chat_bubble_outline, color: Colors.white, size: 20),
                            ),
                            const SizedBox(width: 8.0),
                          ],
                          Flexible(
                            child: Container(
                              constraints: BoxConstraints(
                                maxWidth: MediaQuery.of(context).size.width * 0.75,
                              ),
                              padding: const EdgeInsets.all(12.0),
                              decoration: BoxDecoration(
                                color: isUser
                                    ? Theme.of(context).colorScheme.primary
                                    : Theme.of(context).colorScheme.secondary.withOpacity(0.2),
                                borderRadius: BorderRadius.only(
                                  topLeft: const Radius.circular(15.0),
                                  topRight: const Radius.circular(15.0),
                                  bottomLeft: isUser
                                      ? const Radius.circular(15.0)
                                      : const Radius.circular(5.0),
                                  bottomRight: isUser
                                      ? const Radius.circular(5.0)
                                      : const Radius.circular(15.0),
                                ),
                              ),
                              child: Text(
                                message.text,
                                style: TextStyle(
                                  color: isUser ? Colors.white : Theme.of(context).textTheme.bodyLarge?.color,
                                ),
                              ),
                            ),
                          ),
                          if (isUser) ...[
                            const SizedBox(width: 8.0),
                            CircleAvatar(
                              backgroundColor: Theme.of(context).primaryColor,
                              child: const Icon(Icons.person, color: Colors.white, size: 20),
                            ),
                          ],
                        ],
                      ),
                    );
                  },
                );
              },
            ),
          ),
          if (Provider.of<ChatbotProvider>(context).isLoading)
            const Padding(
              padding: EdgeInsets.all(8.0),
              child: LinearProgressIndicator(),
            ),
          Padding(
            padding: const EdgeInsets.all(8.0),
            child: Row(
              children: <Widget>[
                Expanded(
                  child: TextField(
                    controller: _textController,
                    decoration: InputDecoration(
                      hintText: 'Enter your message...',
                    ).applyDefaults(Theme.of(context).inputDecorationTheme),
                    onSubmitted: (_) => _sendMessage(),
                  ),
                ),
                const SizedBox(width: 8.0),
                FloatingActionButton(
                  onPressed: _sendMessage,
                  backgroundColor: Theme.of(context).colorScheme.secondary,
                  foregroundColor: Theme.of(context).colorScheme.onSecondary,
                  child: const Icon(Icons.send),
                ),
              ],
            ),
          ),
        ],
      ),
    );
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ChatbotPage at {file_path} with enhanced message bubble aesthetics.")

# --- 9. Re-create image_generation_service.dart ---
file_path = os.path.join('lib', 'domain', 'usecases', 'image_generation_service.dart')
file_content = '''
import 'package:ai_chatbot_app/data/datasources/ai_data_source.dart';

// Abstract interface for text-to-image generation
abstract class ImageGenerationService {
  Future<List<int>> generateImage(String prompt);
}

// Concrete implementation of ImageGenerationService
class ImageGenerationServiceImpl implements ImageGenerationService {
  final AiDataSource _aiDataSource;

  ImageGenerationServiceImpl(this._aiDataSource);

  @override
  Future<List<int>> generateImage(String prompt) async {
    return await _aiDataSource.generateImageFromText(prompt);
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ImageGenerationService at {file_path}")

# --- 10. Re-create image_generation_provider.dart ---
file_path = os.path.join('lib', 'presentation', 'providers', 'image_generation_provider.dart')
file_content = '''
import 'package:flutter/material.dart';
import 'package:ai_chatbot_app/domain/usecases/image_generation_service.dart';

class ImageGenerationProvider with ChangeNotifier {
  final ImageGenerationService _imageGenerationService;
  List<int>? _generatedImageBytes;
  bool _isLoading = false;
  String? _errorMessage;

  ImageGenerationProvider(this._imageGenerationService);

  List<int>? get generatedImageBytes => _generatedImageBytes;
  bool get isLoading => _isLoading;
  String? get errorMessage => _errorMessage;

  Future<void> generateImage(String prompt) async {
    if (prompt.trim().isEmpty) {
      _errorMessage = 'Please enter a prompt to generate an image.';
      notifyListeners();
      return;
    }

    _isLoading = true;
    _errorMessage = null;
    _generatedImageBytes = null;
    notifyListeners();

    try {
      _generatedImageBytes = await _imageGenerationService.generateImage(prompt);
    } catch (e) {
      _errorMessage = 'Error generating image: $e';
      print('ImageGenerationProvider error: $e');
    }

    _isLoading = false;
    notifyListeners();
  }

  void clearImage() {
    _generatedImageBytes = null;
    _errorMessage = null;
    notifyListeners();
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ImageGenerationProvider at {file_path}")

# --- 11. Re-create and fix image_generation_page.dart ---
file_path = os.path.join('lib', 'presentation', 'pages', 'image_generation_page.dart')
file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/presentation/providers/image_generation_provider.dart';
import 'dart:typed_data'; // Required for Uint8List.fromList

class ImageGenerationPage extends StatefulWidget {
  const ImageGenerationPage({super.key});

  @override
  State<ImageGenerationPage> createState() => _ImageGenerationPageState();
}

class _ImageGenerationPageState extends State<ImageGenerationPage> {
  final TextEditingController _promptController = TextEditingController();

  @override
  void dispose() {
    _promptController.dispose();
    super.dispose();
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        title: const Text('Text-to-Image Generator'),
      ),
      body: Consumer<ImageGenerationProvider>(
        builder: (context, imageProvider, child) {
          return SingleChildScrollView(
            padding: const EdgeInsets.all(16.0),
            child: Column(
              crossAxisAlignment: CrossAxisAlignment.stretch,
              children: <Widget>[
                TextField(
                  controller: _promptController,
                  decoration: InputDecoration(
                    hintText: 'Enter your image prompt (e.g., "a cat in space")',
                  ).applyDefaults(Theme.of(context).inputDecorationTheme),
                  maxLines: 3,
                ),
                const SizedBox(height: 16.0),

                imageProvider.isLoading
                    ? Center(
                        child: Column(
                          children: [
                            const CircularProgressIndicator(),
                            const SizedBox(height: 8.0),
                            Text('Generating image...', style: Theme.of(context).textTheme.bodyMedium),
                          ],
                        ),
                      )
                    : ElevatedButton(
                        onPressed: () {
                          imageProvider.generateImage(_promptController.text);
                        },
                        child: const Text('Generate Image'),
                      ),
                const SizedBox(height: 16.0),

                if (imageProvider.errorMessage != null)
                  Padding(
                    padding: const EdgeInsets.only(bottom: 16.0),
                    child: Text(
                      imageProvider.errorMessage!,
                      style: Theme.of(context).textTheme.bodyMedium?.copyWith(color: Colors.red),
                      textAlign: TextAlign.center,
                    ),
                  ),

                if (imageProvider.generatedImageBytes != null)
                  Column(
                    children: [
                      const Text('Generated Image:', style: TextStyle(fontWeight: FontWeight.bold)),
                      const SizedBox(height: 8.0),
                      Container(
                        decoration: BoxDecoration(
                          border: Border.all(color: Theme.of(context).colorScheme.primary.withOpacity(0.5), width: 2),
                          borderRadius: BorderRadius.circular(10),
                        ),
                        child: ClipRRect(
                          borderRadius: BorderRadius.circular(8),
                          child: Image.memory(
                            Uint8List.fromList(imageProvider.generatedImageBytes!),
                            fit: BoxFit.contain,
                            errorBuilder: (context, error, stackTrace) => const Icon(Icons.error_outline, color: Colors.red, size: 50),
                          ),
                        ),
                      ),
                      const SizedBox(height: 16.0),
                      ElevatedButton.icon(
                        onPressed: imageProvider.clearImage,
                        icon: const Icon(Icons.clear),
                        label: const Text('Clear Image'),
                      ),
                    ],
                  )
                else if (!imageProvider.isLoading)
                  Padding(
                    padding: const EdgeInsets.all(24.0),
                    child: Text(
                      """No image generated yet.\nEnter a prompt and click "Generate Image""",
                      textAlign: TextAlign.center,
                      style: Theme.of(context).textTheme.bodyMedium?.copyWith(color: Colors.grey[600]),
                    ),
                  ),
              ],
            ),
          );
        },
      ),
    );
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ImageGenerationPage at {file_path} with improved UI and visual cues.")

# --- 12. Re-create image_modification_service.dart ---
file_path = os.path.join('lib', 'domain', 'usecases', 'image_modification_service.dart')
file_content = '''
import 'package:ai_chatbot_app/data/datasources/ai_data_source.dart';

// Abstract interface for image modification
abstract class ImageModificationService {
  Future<List<int>> modifyImage(List<int> imageData, String prompt);
}

// Concrete implementation of ImageModificationService
class ImageModificationServiceImpl implements ImageModificationService {
  final AiDataSource _aiDataSource;

  ImageModificationServiceImpl(this._aiDataSource);

  @override
  Future<List<int>> modifyImage(List<int> imageData, String prompt) async {
    return await _aiDataSource.modifyImage(imageData, prompt);
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ImageModificationService at {file_path}")

# --- 13. Re-create image_modification_provider.dart ---
file_path = os.path.join('lib', 'presentation', 'providers', 'image_modification_provider.dart')
file_content = '''
import 'package:flutter/material.dart';
import 'package:image_picker/image_picker.dart';
import 'dart:io';
import 'package:ai_chatbot_app/domain/usecases/image_modification_service.dart';

class ImageModificationProvider with ChangeNotifier {
  final ImageModificationService _imageModificationService;

  File? _originalImageFile;
  List<int>? _originalImageBytes;
  String _prompt = '';
  List<int>? _modifiedImageBytes;
  bool _isLoading = false;
  String? _errorMessage;

  ImageModificationProvider(this._imageModificationService);

  File? get originalImageFile => _originalImageFile;
  List<int>? get originalImageBytes => _originalImageBytes;
  String get prompt => _prompt;
  List<int>? get modifiedImageBytes => _modifiedImageBytes;
  bool get isLoading => _isLoading;
  String? get errorMessage => _errorMessage;

  void setPrompt(String newPrompt) {
    _prompt = newPrompt;
    notifyListeners();
  }

  Future<void> pickImage() async {
    final ImagePicker picker = ImagePicker();
    final XFile? image = await picker.pickImage(source: ImageSource.gallery);

    if (image != null) {
      _originalImageFile = File(image.path);
      _originalImageBytes = await _originalImageFile!.readAsBytes();
      _modifiedImageBytes = null;
      _errorMessage = null;
      notifyListeners();
    }
  }

  Future<void> modifyImage() async {
    if (_originalImageBytes == null) {
      _errorMessage = 'Please pick an image first.';
      notifyListeners();
      return;
    }
    if (_prompt.trim().isEmpty) {
      _errorMessage = 'Please enter a prompt for modification.';
      notifyListeners();
      return;
    }

    _isLoading = true;
    _errorMessage = null;
    _modifiedImageBytes = null;
    notifyListeners();

    try {
      _modifiedImageBytes = await _imageModificationService.modifyImage(
          _originalImageBytes!, _prompt);
    } catch (e) {
      _errorMessage = 'Error modifying image: $e';
      print('ImageModificationProvider error: $e');
    }

    _isLoading = false;
    notifyListeners();
  }

  void clearAll() {
    _originalImageFile = null;
    _originalImageBytes = null;
    _prompt = '';
    _modifiedImageBytes = null;
    _errorMessage = null;
    notifyListeners();
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ImageModificationProvider at {file_path}")

# --- 14. Re-create and fix image_modification_page.dart ---
file_path = os.path.join('lib', 'presentation', 'pages', 'image_modification_page.dart')
file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/presentation/providers/image_modification_provider.dart';
import 'dart:typed_data'; // Required for Uint8List.fromList

class ImageModificationPage extends StatefulWidget {
  const ImageModificationPage({super.key});

  @override
  State<ImageModificationPage> createState() => _ImageModificationPageState();
}

class _ImageModificationPageState extends State<ImageModificationPage> {
  final TextEditingController _promptController = TextEditingController();

  @override
  void initState() {
    super.initState();
    _promptController.addListener(() {
      Provider.of<ImageModificationProvider>(context, listen: false)
          .setPrompt(_promptController.text);
    });
  }

  @override
  void dispose() {
    _promptController.dispose();
    super.dispose();
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(
        title: const Text('Image Modifier'),
        actions: [
          IconButton(
            icon: const Icon(Icons.clear_all),
            onPressed: () {
              Provider.of<ImageModificationProvider>(context, listen: false).clearAll();
              _promptController.clear();
            },
          ),
        ],
      ),
      body: Consumer<ImageModificationProvider>(
        builder: (context, imageModifierProvider, child) {
          return SingleChildScrollView(
            padding: const EdgeInsets.all(16.0),
            child: Column(
              crossAxisAlignment: CrossAxisAlignment.stretch,
              children: <Widget>[
                ElevatedButton.icon(
                  onPressed: imageModifierProvider.pickImage,
                  icon: const Icon(Icons.image),
                  label: const Text('Pick Image to Modify'),
                  style: Theme.of(context).elevatedButtonTheme.style?.copyWith(
                    backgroundColor: MaterialStateProperty.all(Theme.of(context).primaryColor),
                    foregroundColor: MaterialStateProperty.all(Colors.white),
                  ),
                ),
                const SizedBox(height: 16.0),

                TextField(
                  controller: _promptController,
                  decoration: InputDecoration(
                    hintText: 'Enter modification prompt (e.g., "make it vintage", "add sunglasses")',
                  ).applyDefaults(Theme.of(context).inputDecorationTheme),
                  maxLines: 3,
                ),
                const SizedBox(height: 16.0),

                imageModifierProvider.isLoading
                    ? Center(
                        child: Column(
                          children: [
                            const CircularProgressIndicator(),
                            const SizedBox(height: 8.0),
                            Text('Modifying image...', style: Theme.of(context).textTheme.bodyMedium),
                          ],
                        ),
                      )
                    : ElevatedButton.icon(
                        onPressed: imageModifierProvider.originalImageBytes != null && _promptController.text.isNotEmpty
                            ? imageModifierProvider.modifyImage
                            : null,
                        icon: const Icon(Icons.auto_fix_high),
                        label: const Text('Apply Modification'),
                      ),
                const SizedBox(height: 16.0),

                if (imageModifierProvider.errorMessage != null)
                  Padding(
                    padding: const EdgeInsets.only(bottom: 16.0),
                    child: Text(
                      imageModifierProvider.errorMessage!,
                      style: Theme.of(context).textTheme.bodyMedium?.copyWith(color: Colors.red),
                      textAlign: TextAlign.center,
                    ),
                  ),

                if (imageModifierProvider.originalImageBytes != null)
                  Column(
                    crossAxisAlignment: CrossAxisAlignment.stretch,
                    children: [
                      Text('Original Image:', style: Theme.of(context).textTheme.headlineSmall?.copyWith(fontWeight: FontWeight.bold)),
                      const SizedBox(height: 8.0),
                      Container(
                        decoration: BoxDecoration(
                          border: Border.all(color: Theme.of(context).colorScheme.primary.withOpacity(0.5), width: 2),
                          borderRadius: BorderRadius.circular(10),
                        ),
                        child: ClipRRect(
                          borderRadius: BorderRadius.circular(8),
                          child: Image.memory(Uint8List.fromList(imageModifierProvider.originalImageBytes!),
                              height: 200, fit: BoxFit.contain),
                        ),
                      ),
                      const SizedBox(height: 16.0),
                    ],
                  ),

                if (imageModifierProvider.modifiedImageBytes != null)
                  Column(
                    crossAxisAlignment: CrossAxisAlignment.stretch,
                    children: [
                      Text('Modified Image:', style: Theme.of(context).textTheme.headlineSmall?.copyWith(fontWeight: FontWeight.bold)),
                      const SizedBox(height: 8.0),
                      Container(
                        decoration: BoxDecoration(
                          border: Border.all(color: Theme.of(context).colorScheme.secondary.withOpacity(0.5), width: 2),
                          borderRadius: BorderRadius.circular(10),
                        ),
                        child: ClipRRect(
                          borderRadius: BorderRadius.circular(8),
                          child: Image.memory(Uint8List.fromList(imageModifierProvider.modifiedImageBytes!),
                              height: 200, fit: BoxFit.contain),
                        ),
                      ),
                      const SizedBox(height: 16.0),
                    ],
                  ),
              ],
            ),
          );
        },
      ),
    );
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created ImageModificationPage at {file_path} with improved UI and visual cues.")

# --- 15. Re-create main.dart (including theme and navigation) ---
file_path = os.path.join('lib', 'main.dart')
file_content = '''
import 'package:flutter/material.dart';
import 'package:provider/provider.dart';
import 'package:ai_chatbot_app/data/datasources/ai_data_source.dart';
import 'package:ai_chatbot_app/data/datasources/tflite_ai_data_source.dart';
import 'package:ai_chatbot_app/domain/usecases/chatbot_service.dart';
import 'package:ai_chatbot_app/domain/usecases/image_generation_service.dart';
import 'package:ai_chatbot_app/domain/usecases/image_modification_service.dart';
import 'package:ai_chatbot_app/presentation/pages/chatbot_page.dart';
import 'package:ai_chatbot_app/presentation/pages/image_generation_page.dart';
import 'package:ai_chatbot_app/presentation/pages/image_modification_page.dart';
import 'package:ai_chatbot_app/presentation/providers/chatbot_provider.dart';
import 'package:ai_chatbot_app/presentation/providers/image_generation_provider.dart';
import 'package:ai_chatbot_app/presentation/providers/image_modification_provider.dart';

void main() async {
  WidgetsFlutterBinding.ensureInitialized();

  final AiDataSource aiDataSource = TfliteAiDataSource();
  await (aiDataSource as TfliteAiDataSource).init(); // Initialize models

  final ChatbotService chatbotService = ChatbotServiceImpl(aiDataSource);
  final ImageGenerationService imageGenerationService = ImageGenerationServiceImpl(aiDataSource);
  final ImageModificationService imageModificationService = ImageModificationServiceImpl(aiDataSource);

  runApp(
    MultiProvider(
      providers: [
        ChangeNotifierProvider(
          create: (_) => ChatbotProvider(chatbotService),
        ),
        ChangeNotifierProvider(
          create: (_) => ImageGenerationProvider(imageGenerationService),
        ),
        ChangeNotifierProvider(
          create: (_) => ImageModificationProvider(imageModificationService),
        ),
      ],
      child: const MyApp(),
    ),
  );
}

class MyApp extends StatelessWidget {
  const MyApp({super.key});

  @override
  Widget build(BuildContext context) {
    return MaterialApp(
      title: 'AI Chatbot App',
      theme: ThemeData(
        primarySwatch: Colors.blueGrey,
        primaryColor: Colors.blueGrey[700],
        colorScheme: ColorScheme.fromSwatch(
          primarySwatch: Colors.blueGrey,
        ).copyWith(
          secondary: Colors.tealAccent[400],
        ),
        scaffoldBackgroundColor: Colors.grey[50],
        appBarTheme: AppBarTheme(
          backgroundColor: Colors.blueGrey[800],
          foregroundColor: Colors.white,
          elevation: 0,
          centerTitle: true,
          titleTextStyle: TextStyle(
            color: Colors.white,
            fontSize: 20,
            fontWeight: FontWeight.bold,
          ),
        ),
        elevatedButtonTheme: ElevatedButtonThemeData(
          style: ElevatedButton.styleFrom(
            backgroundColor: Colors.tealAccent[400],
            foregroundColor: Colors.blueGrey[900],
            padding: EdgeInsets.symmetric(horizontal: 20, vertical: 15),
            shape: RoundedRectangleBorder(
              borderRadius: BorderRadius.circular(10),
            ),
          ),
        ),
        inputDecorationTheme: InputDecorationTheme(
          border: OutlineInputBorder(
            borderRadius: BorderRadius.circular(10),
            borderSide: BorderSide.none,
          ),
          filled: true,
          fillColor: Colors.grey[200],
          contentPadding: EdgeInsets.symmetric(horizontal: 16, vertical: 12),
          hintStyle: TextStyle(color: Colors.grey[600]),
        ),
        textTheme: TextTheme(
          headlineMedium: TextStyle(color: Colors.blueGrey[800], fontWeight: FontWeight.bold),
          bodyLarge: TextStyle(color: Colors.blueGrey[700]),
          bodyMedium: TextStyle(color: Colors.blueGrey[600]),
          labelLarge: TextStyle(color: Colors.blueGrey[900], fontWeight: FontWeight.bold),
        ),
        visualDensity: VisualDensity.adaptivePlatformDensity,
      ),
      home: const MainScreen(),
    );
  }
}

class MainScreen extends StatefulWidget {
  const MainScreen({super.key});

  @override
  State<MainScreen> createState() => _MainScreenState();
}

class _MainScreenState extends State<MainScreen> {
  int _selectedIndex = 0;

  static const List<Widget> _widgetOptions = <Widget>[
    ChatbotPage(),
    ImageGenerationPage(),
    ImageModificationPage(),
  ];

  void _onItemTapped(int index) {
    setState(() {
      _selectedIndex = index;
    });
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      body: Center(
        child: _widgetOptions.elementAt(_selectedIndex),
      ),
      bottomNavigationBar: BottomNavigationBar(
        items: const <BottomNavigationBarItem>[
          BottomNavigationBarItem(
            icon: Icon(Icons.chat),
            label: 'Chatbot',
          ),
          BottomNavigationBarItem(
            icon: Icon(Icons.image),
            label: 'Image Gen',
          ),
          BottomNavigationBarItem(
            icon: Icon(Icons.edit),
            label: 'Image Modify',
          ),
        ],
        currentIndex: _selectedIndex,
        selectedItemColor: Colors.tealAccent[400],
        unselectedItemColor: Colors.blueGrey[400],
        backgroundColor: Colors.blueGrey[800],
        onTap: _onItemTapped,
      ),
    );
  }
}
'''
with open(file_path, 'w') as f:
    f.write(file_content)
print(f"Re-created main.dart to integrate all features and theme at {file_path}")

# --- 16. Update pubspec.yaml with dependencies ---
pubspec_path = 'pubspec.yaml'
with open(pubspec_path, 'r') as f:
    pubspec_content = f.readlines()

updated_content = []
inserted_deps = False
for line in pubspec_content:
    updated_content.append(line)
    # Check for the flutter sdk dependency line and insert other dependencies after it
    if 'sdk: flutter' in line and not inserted_deps:
        # Ensure correct indentation for newly inserted dependencies
        updated_content.append(
            "  tflite_flutter: ^0.12.1\n"
            "  provider: ^6.1.1\n"
            "  image_picker: ^1.0.7\n"
            "  image: ^4.1.7\n"
            "  path_provider: ^2.1.2\n"
        )
        inserted_deps = True

# Also ensure assets/models/ is declared
# This logic is made more robust to ensure assets: block is added correctly or updated.
final_pubspec_content = []
flutter_section_start = -1
assets_section_exists = False

for i, line in enumerate(updated_content):
    final_pubspec_content.append(line)
    if line.strip() == 'flutter:':
        flutter_section_start = i
    if flutter_section_start != -1 and line.strip() == 'assets:':
        assets_section_exists = True

if flutter_section_start != -1 and not assets_section_exists:
    # Find the end of the flutter section to insert assets before the next top-level key
    insert_index = flutter_section_start + 1
    while insert_index < len(final_pubspec_content) and final_pubspec_content[insert_index].startswith(' '):
        insert_index += 1

    # Insert assets section
    final_pubspec_content.insert(insert_index, '  assets:\n')
    final_pubspec_content.insert(insert_index + 1, '    - assets/models/\n')
elif assets_section_exists:
    # If assets section already exists, check if assets/models/ is present, and add if not
    models_asset_present = False
    for i, line in enumerate(final_pubspec_content):
        if line.strip() == 'assets:':
            # Check lines below 'assets:' for '    - assets/models/'
            for j in range(i + 1, len(final_pubspec_content)):
                if not final_pubspec_content[j].startswith('    -') and final_pubspec_content[j].strip() != '':
                    break # End of assets block
                if 'assets/models/' in final_pubspec_content[j]:
                    models_asset_present = True
                    break
        if models_asset_present: break

    if not models_asset_present:
        # Find the end of the assets block to insert '    - assets/models/'
        insert_idx = -1
        for i, line in enumerate(final_pubspec_content):
            if line.strip() == 'assets:':
                insert_idx = i + 1
                while insert_idx < len(final_pubspec_content) and final_pubspec_content[insert_idx].startswith('    -'):
                    insert_idx += 1
                break
        if insert_idx != -1:
            final_pubspec_content.insert(insert_idx, '    - assets/models/\n')

with open(pubspec_path, 'w') as f:
    f.writelines(final_pubspec_content)
print("Dependencies and assets re-added to pubspec.yaml.")

# --- 17. Run flutter pub get ---
print("Running flutter pub get...")
!flutter pub get
print("flutter pub get completed.")

# --- 18. Fix build.gradle.kts to specify SDK versions ---
# The build.gradle.kts might not be generated in every project, so check existence first.
build_gradle_kts_path = os.path.join('android', 'app', 'build.gradle.kts')

if os.path.exists(build_gradle_kts_path):
    print(f"Modifying {build_gradle_kts_path} to set compileSdkVersion, targetSdkVersion, and buildToolsVersion to 33...")
    with open(build_gradle_kts_path, 'r') as f:
        content = f.readlines()

    updated_content_gradle = []
    for line in content:
        if 'compileSdkVersion' in line:
            updated_content_gradle.append("    compileSdkVersion = 33\n")
        elif 'buildToolsVersion' in line:
            updated_content_gradle.append('    buildToolsVersion = "33.0.0"\n')
        elif 'targetSdk' in line and 'flutter.targetSdkVersion' in line:
            updated_content_gradle.append("        targetSdk = 33\n")
        else:
            updated_content_gradle.append(line)

    with open(build_gradle_kts_path, 'w') as f:
        f.writelines(updated_content_gradle)
    print(f"Updated {build_gradle_kts_path} with compileSdkVersion = 33, targetSdk = 33, and buildToolsVersion = \"33.0.0\".")
else:
    print(f"Warning: {build_gradle_kts_path} not found. Skipping build.gradle.kts modification.")

# --- 19. Run flutter clean ---
print("Running flutter clean...")
!flutter clean
print("flutter clean completed.")

# --- 20. Run flutter doctor -v to verify Android SDK setup ---
print("Verifying Android SDK setup with flutter doctor -v...")
!flutter doctor -v

# --- 21. Run flutter build apk --analyze-size --target-platform=android-arm64 ---
print("Running flutter build apk --analyze-size --target-platform=android-arm64...")
!flutter build apk --analyze-size --target-platform=android-arm64

# Navigate back to the original directory
os.chdir('..')
print(f"Navigated back to: {os.getcwd()}")
````
